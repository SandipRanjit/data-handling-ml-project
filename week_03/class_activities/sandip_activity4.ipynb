# Task 3: Convert an Existing Dataset

```python
def convert_example(example):
    return {
        "messages": [
            {
                "role": "user",
                "content": example["question"]
            },
            {
                "role": "assistant",
                "content": example["answer"]
            }
        ]
    }
```

```python
def convert_example(example):
    # Handle potential missing values or special characters
    weight = example.get('weight', 'N/A')
    if weight == '?':
        weight = 'Unknown'

    max_glu_serum = example.get('max_glu_serum', 'N/A')
    if max_glu_serum is None:
        max_glu_serum = 'N/A'

    a1c_result = example.get('A1Cresult', 'N/A')
    if a1c_result is None:
        a1c_result = 'N/A'

    # Construct the user's message using the identified input fields
    user_content = (
        f"Given the patient's characteristics: "
        f"Race: {example.get('race', 'N/A')}, "
        f"Gender: {example.get('gender', 'N/A')}, "
        f"Age: {example.get('age', 'N/A')}, "
        f"Weight: {weight}, "
        f"Max Glucose Serum: {max_glu_serum}, "
        f"A1C Result: {a1c_result}, "
        f"Metformin: {example.get('metformin', 'N/A')}, "
        f"Glipizide: {example.get('glipizide', 'N/A')}, "
        f"Insulin: {example.get('insulin', 'N/A')}, "
        f"Change: {example.get('change', 'N/A')}, "
        f"Diabetes Medication: {example.get('diabetesMed', 'N/A')}. "
        f"Predict if the patient will be readmitted to the hospital."
    )

    # Construct the assistant's message using the identified target output field
    assistant_content = str(example.get('readmitted', 'N/A'))

    return {
        "messages": [
            {
                "role": "user",
                "content": user_content
            },
            {
                "role": "assistant",
                "content": assistant_content
            }
        ]
    }

formatted_dataset = dataset.map(convert_example)
```

# Task 4: Save the Dataset as JSONL

```python
import json

with open("train.jsonl", "w", encoding="utf-8") as f:
    for example in formatted_dataset["train"]:
        f.write(
            json.dumps(example, ensure_ascii=False) + "\n"
        )
```
