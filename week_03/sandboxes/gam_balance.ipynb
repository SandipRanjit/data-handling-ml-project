```python
# Create a new column called "balance_diff"
# It checks whether:
# Total Assets = Total Liabilities + Total Equity

balance_clean["balance_diff"] = (
    balance_clean["Total Assets"]
    - (
        balance_clean["Total Liabilities"]
        + balance_clean["Total Equity"]
    )
)
```

```python
# Check the balance using SimFin's complete balance-sheet total
df["balance_check"] = (
    df["Total Assets"]
    - df["Total Liabilities & Equity"]
)

df["balance_check"].abs().describe()
```

```python
# Show the 10 rows with the largest mismatch
df[
    [
        "Ticker",
        "Fiscal Year",
        "Total Assets",
        "Total Liabilities & Equity",
        "balance_check"
    ]
].sort_values(
    "balance_check",
    key=abs,
    ascending=False
).head(10)
```

```python
# Accounting mismatch as a percentage of company assets
balance_clean["balance_error_pct"] = (
    balance_clean["balance_diff"].abs()
    / balance_clean["Total Assets"]
) * 100

balance_clean["balance_error_pct"].describe()
```

# Balance Sheet Data Cleaning Summary

## Data Summary

The dataset contains annual balance sheet data for US companies from SimFin. Each row represents one company for one fiscal year.

The main features used include:

- Ticker
- SimFinId
- Fiscal Year
- Cash, Cash Equivalents & Short Term Investments
- Total Current Assets
- Total Assets
- Total Current Liabilities
- Total Liabilities
- Total Equity

## Cleaning Job Summary

First inspected the dataset for missing values, duplicates, data types, and unusual values.

After that, selected the most reliable balance sheet features and removed rows with missing values in essential columns.

Features with a large amount of missing data, such as short-term debt and long-term debt, were excluded from the cleaned dataset rather than filling them with artificial values.

Aalso checked the accounting consistency between **Total Assets** and **Total Liabilities & Equity**.

Seven rows with a mismatch greater than **1%** were identified as likely data-quality errors and removed.

Finally, visualized the data to inspect its distribution and relationships between financial features.

## Summary

- SimFin annual US company balance sheet data
- Selected reliable financial features
- Checked missing values and duplicates
- Removed rows missing essential information
- Avoided unnecessary imputation
- Validated balance sheet consistency
- Removed 7 abnormal records with more than 1% accounting mismatch
- Visualized asset distribution and financial relationships
- Prepared cleaned data for merging with other financial tables
