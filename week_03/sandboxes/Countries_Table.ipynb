```python
import pandas as pd

read_df = pd.read_csv('/content/drive/MyDrive/Data Handling and Machine Learning/week_03/raw/company_info.csv', sep=';')
display(read_df.head())
```

```python
df_copy = read_df.copy()
```

```python
df_copy.drop(columns=['IndustryId', 'ISIN', 'End of financial year (month)', 'Number Employees', 'Business Summary','Market','CIK'], inplace=True)

display(df_copy.head())
```

```python
companies_excel_df = pd.read_excel('/content/drive/MyDrive/Data Handling and Machine Learning/week_03/raw/100_companies.xlsx')

display(companies_excel_df.head())
```

```python
excel_simfin_ids = companies_excel_df['SimFinId'].unique()
selected_companies_df = df_copy[df_copy['SimFinId'].isin(excel_simfin_ids)]

display(selected_companies_df.head())
```

```python
output_filename = '/content/selected_companies.csv'
selected_companies_df.to_csv(output_filename, index=False)
print(f"DataFrame successfully exported to '{output_filename}'")
```
