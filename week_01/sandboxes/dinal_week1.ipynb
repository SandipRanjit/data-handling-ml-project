## Overview of Google Colab and Cell Types

Google Colab is an interactive, browser-based cloud environment based on Jupyter Notebooks. It allows executable code, data visualizations, and structured text to coexist in a single document while providing access to cloud-hosted CPU and GPU runtimes.

---

#### Primary Cell Types

1. **Code Cells**
   * **Function:** Executable cells used to run code, load datasets, and perform matrix calculations.
   * **Behavior:** Code is sent directly to the active **Python Kernel** running on the cloud virtual machine. Any variables, functions, or imports created in a code cell remain saved in memory (RAM/VRAM) and can be accessed by subsequent cells. Outputs—such as printed text, terminal logs, or plotted charts—appear directly beneath the cell.

2. **Text (Markdown) Cells**
   * **Function:** Static cells used for documenting methodology, providing theoretical context, and structuring the report.
   * **Behavior:** These cells do not execute Python code. Instead, they render standard **Markdown** (for headings, bulleted lists, and tables) alongside **LaTeX** for formatting mathematical models (e.g., $C_{ij} = \sum_{k=1}^{n} A_{ik} B_{kj}$).

---

#### Execution Modes Within Code Cells

* **Standard Python:** Normal Python statements (e.g., `import torch`, `c = torch.matmul(a, b)`) execute within the persistent Python process.
* **Shell Commands (`!`):** Prefixing a command with an exclamation point (e.g., `!nvidia-smi` or `!ls`) executes instructions in a temporary system terminal shell outside the Python environment.
* **Magic Commands (`%` or `%%`):** Instructions starting with `%` (line magic) or `%%` (cell magic) control the notebook environment directly, such as changing working directories (`%cd`) or timing cell execution (`%timeit`).

# Task 6 - CPU vs GPU Matrix Multiplication

## 6.2 Benchmark implementation

````python
import torch
import time

def benchmark_matmul(n=4000):

    if not torch.cuda.is_available():
        raise RuntimeError("CUDA is not available. Enable a GPU runtime first.")

    print(f"Matrix size: {n} x {n}")
    print(f"GPU: {torch.cuda.get_device_name(0)}")

    # -------------------------
    # CPU
    # -------------------------
    a_cpu = torch.randn(n, n, dtype=torch.float32)
    b_cpu = torch.randn(n, n, dtype=torch.float32)

    # Warm up CPU execution
    _ = torch.matmul(a_cpu, b_cpu)

    start = time.perf_counter()
    c_cpu = torch.matmul(a_cpu, b_cpu)
    cpu_time = time.perf_counter() - start

    # -------------------------
    # GPU
    # -------------------------
    a_gpu = a_cpu.to("cuda")
    b_gpu = b_cpu.to("cuda")

    # Warm up GPU
    _ = torch.matmul(a_gpu, b_gpu)
    torch.cuda.synchronize()

    start = time.perf_counter()

    c_gpu = torch.matmul(a_gpu, b_gpu)

    # GPU operations are asynchronous.
    # Wait for the GPU before stopping the timer.
    torch.cuda.synchronize()

    gpu_time = time.perf_counter() - start

    speedup = cpu_time / gpu_time

    print(f"CPU time: {cpu_time:.4f} s")
    print(f"GPU time: {gpu_time:.4f} s")
    print(f"GPU speedup: {speedup:.2f}x")

    return cpu_time, gpu_time, speedup
````

````python
cpu_time, gpu_time, speedup = benchmark_matmul(4000)
````

## 7.1 Deliberately perform an incorrect benchmark

**Writing the benchmark function as reusable Python module**

````python
%%writefile benchmark_v2.py

import torch
import time

def benchmark_matmul(n=4000, sync=True):

    if not torch.cuda.is_available():
        raise RuntimeError("CUDA is not available. Enable a GPU runtime first.")

    print(f"Matrix size: {n} x {n}")
    print(f"GPU: {torch.cuda.get_device_name(0)}")

    # -------------------------
    # CPU
    # -------------------------
    a_cpu = torch.randn(n, n, dtype=torch.float32)
    b_cpu = torch.randn(n, n, dtype=torch.float32)

    # Warm up CPU execution
    _ = torch.matmul(a_cpu, b_cpu)

    start_cpu = time.perf_counter()
    c_cpu = torch.matmul(a_cpu, b_cpu)
    cpu_time = time.perf_counter() - start_cpu

    # -------------------------
    # GPU
    # -------------------------
    a_gpu = a_cpu.to("cuda")
    b_gpu = b_cpu.to("cuda")

    # Warm up GPU
    _ = torch.matmul(a_gpu, b_gpu)
    if sync:
        torch.cuda.synchronize()

    start_gpu = time.perf_counter()
    c_gpu = torch.matmul(a_gpu, b_gpu)

    if sync:
        torch.cuda.synchronize() # Wait for GPU to finish

    gpu_time = time.perf_counter() - start_gpu

    speedup = cpu_time / gpu_time if gpu_time > 0 else 0.0

    return cpu_time, gpu_time, speedup
````

1. Does GPU speedup increase with problem size?
- Yes. Matrix multiplication operations grow at $O(N^3)$ computational complexity while memory requirements grow at $O(N^2)$. As the matrix size $N$ grows larger, compute time quickly dominates, allowing the GPU’s thousands of parallel cores to scale far better than a CPU.

2. Is the GPU always faster?
- No. For very tiny problem sizes (e.g., $10 \times 10$ matrices), a CPU is often faster than a GPU.

3. Why might a small matrix not benefit as much?
- GPU Under-utilization: A Tesla T4 has 2,560 CUDA cores. Small matrix operations do not provide enough compute work to keep all cores busy, leaving most of the GPU idle.

- Low Arithmetic Intensity: The compute work isn't large enough to hide the delays in fetching data from VRAM.

4. What role might data transfer and kernel-launch overhead play?
- Kernel Launch Overhead: Sending a execution command from the CPU to the GPU takes a fixed amount of setup time (a few microseconds). For tiny inputs, launching the job takes longer than calculating the result.

- PCIe Bus Transfer: Copying data from host RAM over the PCIe bus into GPU memory introduces high latency. If transfer time exceeds computation time, the GPU loses its speed advantage.

## 10.2 Mathematical model
Include:

$$
C_{ij} = \sum_{k=1}^{n} A_{ik}B_{kj}
$$

Explain that standard matrix multiplication requires $\mathcal{O}(n^3)$ arithmetic work.

---

**Answer:**  To understand why matrix multiplication scales the way it does, we can look at the basic math involved. When multiplying two square matrices $A$ and $B$ (both of size $n \times n$) to produce matrix $C$, each individual entry $C_{ij}$ is calculated using the standard dot product formula:
$$
C_{ij} = \sum_{k=1}^{n} A_{ik}B_{kj}
$$

To find the value for a single position in matrix $C$, the computer takes row $i$ from matrix $A$ and column $j$ from matrix $B$. It performs $n$ separate multiplications and then adds those products together using $n - 1$ additions. This comes out to $2n - 1$ total floating-point operations (FLOPs) per output element.

Since the final matrix $C$ contains $n \times n$ (or $n^2$) total elements, this calculation must be repeated for every cell in the grid:

$$\text{Total Operations} = n^2 \cdot (2n - 1) = 2n^3 - n^2$$

In Big-O notation, lower-order terms and constant coefficients are omitted because they become negligible as $n$ grows large. This leaves an algorithmic time complexity of $O(n^3)$.

This cubic relationship explains why matrix operations quickly become computationally expensive. If the matrix dimension $n$ is doubled, the total number of arithmetic operations increases by a factor of eight ($2^3 = 8$). For example, at $n = 4000$, computing the output requires over 127 billion individual math operations. A standard CPU thread processes these sequentially or in small parallel batches, which leads to a sharp increase in run time. A GPU, on the other hand, handles thousands of these independent calculations simultaneously across its parallel architecture.

## 10.4 Explain the speedup

Matrix multiplication shows dramatic performance differences when shifted from a central processing unit (CPU) to a graphics processing unit (GPU). Understanding this performance gap requires examining how modern processors manage parallel workloads, asynchronous task execution, and system overhead.

**1. why GPUs can accelerate matrix multiplication:**

Matrix multiplication is an inherently parallel workload composed of millions of independent dot products. A standard CPU is optimized for low-latency, sequential processing, featuring a small number of high-speed cores backed by complex cache hierarchies. In contrast, a GPU consists of thousands of smaller, highly specialized cores designed for massive throughput. Because calculating each element $C_{ij}$ requires no data dependencies on neighboring cells, the GPU can execute thousands of multiply-add operations simultaneously across its CUDA cores, outpacing the sequential execution capacity of CPU threads.

**2. why GPU operations are often asynchronous and why synchronization is required for accurate timing:**

PyTorch executes GPU commands asynchronously. When calling a CUDA function like `torch.matmul()`, the CPU simply enqueues the operation into a hardware execution stream and immediately regains control to execute the next line of Python code, without waiting for the GPU to finish calculating.

Because of this non-blocking behavior, standard timing functions running on the CPU will only record the fraction of a millisecond it took to queue the task, rather than the time the GPU took to compute the answer. To capture genuine execution latency, an explicit barrier like `torch.cuda.synchronize()` must be called. This forces the CPU to pause execution until the GPU has completed all pending tasks in the queue.

```markdown
> **Important:** GPU benchmarks must account for asynchronous execution. CPU-side timing without synchronization can underestimate the actual GPU computation time.
```

**4. why the GPU may not provide the same speedup for small workloads:**

While GPUs excel at large workloads, they often provide minimal speedup, or even run slower than a CPU, when processing small matrix sizes (such as $n = 500$ or smaller). This happens due to fixed overhead costs:
  - Kernel Launch Overhead: Preparing and dispatching a CUDA kernel from the CPU to the GPU introduces a constant time delay.
  - Data Transfer Latency: Moving input tensors from system RAM across the PCIe bus into GPU VRAM takes finite time.
  - Underutilization: Small matrices do not contain enough individual operations to occupy thousands of GPU cores at once, leaving a large portion of the hardware idle.
  
  For small matrices, these startup and scheduling overheads dominate the total execution duration. The computational advantages of GPU acceleration only become apparent when the workload $O(n^3)$ grows large enough to outweigh the initial setup costs.

In Google Colab, `/content/` and `/content/drive/MyDrive/` are two totally different storage areas. The main difference comes down to speed versus permanence.

The Main Differences

`/content/` (Local Temporary Storage): This is the local SSD drive directly attached to the virtual machine Colab assigns to us. Because it sits right on the machine, reading and writing files here is extremely fast. However, it is temporary. The moment Colab session ends, times out, or disconnects, everything saved in `/content/` is wiped forever.

`/content/drive/MyDrive/` (Google Drive Storage): This is our actual Google Drive account linked to the notebook. Files saved here are permanent and will not disappear when our session ends. The downside is that because it syncs over the network, file access is noticeably slower than the local drive.

Why This Matters for Machine Learning

When we are training models or running benchmarks, picking the wrong folder can break our workflow or slow down our experiments.

Saving Results and Model Weights: If we spend hours training a PyTorch model and save the final weights (model.pt) or our benchmark plots inside `/content/`, we risk losing all our work if Colab suddenly disconnects or resets. We should always save our final outputs, plots, and model checkpoints to `/content/drive/MyDrive/` so they stay safe.

Training Speed and Data Loading: If we try to stream a dataset made up of thousands of small image or text files straight out of /content/drive/MyDrive/ while training, our GPU will spend most of its time sitting idle waiting for files to download over the network. It is much better to upload or unzip our dataset directly into `/content/` so the GPU can load batches at full speed.

---

The Best Way to Handle It

A good rule of thumb is to read input data from `/content/` for maximum training speed, but save final models, graphs, and logs to `/content/drive/MyDrive/` so we don't lose our work.

### Question 2

Why `!cd /tmp` vs `%cd /tmp` behaves differently?

- `!cd /tmp` fails to persist because the `!` launches a brand new subshell process to execute `cd /tmp`. The working directory changes only inside that temporary child process, which immediately closes as soon as the line finishes running. The main Python kernel stays right where it was.

- `%cd /tmp` works because `%cd` is an IPython built-in magic command that directly changes the working directory of the main, running Python process itself.

---

### Question 3

Why `!touch file.txt` creates a persistent file?

Even though the shell subshell created by `!touch` shuts down instantly, `touch` performs an operation directly on the machine's underlying operating system file system (the disk drive). The file is saved to the disk before the subshell exits, so the file remains accessible to any future process running on that machine—including the Python kernel.

---

### Question 4

What is the difference between the Python kernel and a separate process started with `!python3 script.py`?

- **Python Kernel:** The continuous, background Python process that powers the interactive notebook session. It maintains variables, imported libraries, GPU memory allocations, and functions in RAM from one cell to the next.

- `!python3 script.py`: A completely separate, standalone operating system process. It launches a fresh Python instance from scratch, loads `script.py`, executes it from start to finish, and shuts down completely. It cannot access any variables or functions defined in current notebook cells, nor does it pass its internal variables back to the notebook.

---

### Question 5

Why is `torch.cuda.synchronize()` important when measuring GPU execution time?

PyTorch handles GPU execution asynchronously. When we run a GPU command like `torch.matmul()`, the CPU simply places the instruction onto a GPU hardware queue and immediately moves on to the next line of code without waiting for the GPU to actually do the math.

Without calling `torch.cuda.synchronize()`, CPU timers like `time.perf_counter()` only measure the tiny fraction of a millisecond it took to queue the task, not the actual time the GPU spent calculating the result. Calling `torch.cuda.synchronize()` forces the CPU to pause until the GPU hardware finishes all pending work, giving us an accurate measurement.

In my experiment, the GPU time was measured as only 0.0001320 s without synchronization, compared with 0.0455704 s with synchronization. This demonstrates why synchronization is necessary for accurate GPU timing.

---

### Question 6

Why is a GPU not automatically faster than a CPU for every computation?

GPUs are designed for throughput (doing thousands of simple, independent tasks at the exact same time), whereas CPUs are designed for low latency (doing complex, sequential tasks step-by-step as fast as possible).

A GPU will be slower than a CPU when:

1. The workload is small: Moving data over to the GPU and launching a CUDA kernel takes setup time (overhead). For small operations, that setup time takes longer than the actual math.

2. The code is sequential: If step B strictly requires the answer from step A, the thousands of GPU cores sit idle because the task cannot be split up in parallel.

The `/content/` directory lives on a temporary, local virtual machine drive allocated to us by Google Colab. As soon as the session times out, disconnects due to inactivity, or encounters a runtime crash, the virtual machine is destroyed—and all data inside `/content/` is permanently deleted.
