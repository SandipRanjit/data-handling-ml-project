```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/utils.py

import os
import base64
from pathlib import Path

ASSET_PATH = ("/content/drive/MyDrive/data_handling_and_machine_learning/UI/assets/")

GREEN = "#6FCF97"
PURPLE = "#A05ACB"
BLUE = "#00D6FF"
YELLOW = "#f7ce46"
BLUE_PURPLE = "#7B7EDB"

def load_svg(filename):
    path = os.path.join(ASSET_PATH, filename)

    with open(path, "r") as f:
        return f.read()

def load_svg_as_image(filename):

    svg = load_svg(filename)

    svg = svg.replace(
        "currentColor",
        "#B86CFF"
    )

    encoded = base64.b64encode(
        svg.encode("utf-8")
    ).decode("utf-8")

    return f"data:image/svg+xml;base64,{encoded}"
```

# Chat

Fixes done:
- Update the chat input to use a Streamlit form so the question field clears automatically after sending.

- Connect the InvestIQ AI Methods to Chat

- The chat will use the selected analysis method:

  - Base Qwen → `base_qwen.generate_response()`
  - Fine-Tuned Qwen → `finetuned_qwen.generate_response()`
  - RAG → `rag.generate_response()`

Models remain lazy-loaded, so selecting a method does not load all three models into GPU memory at once.

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/chat.py

import streamlit as st
import json
import markdown
from modules import utils
from modules import base_qwen
from modules import finetuned_qwen
from modules import rag


def initialize_chat():

    if "messages" not in st.session_state:
        st.session_state.messages = [
            {
                "role": "assistant",
                "message": (
                    "Hi! I'm your InvestIQ research assistant. "
                    "Ask me anything about the analyzed companies, "
                    "risk, sectors, or your investment strategy."
                )
            }
        ]

    if "chat_method" not in st.session_state:
        st.session_state.chat_method = "Base Qwen"

def normalize_ai_response(response):
    """
    Convert model output into plain human-readable text
    for the Streamlit chat UI.
    """

    if not isinstance(response, str):
        return str(response)

    response = response.strip()

    try:
        parsed = json.loads(response)

        if isinstance(parsed, dict):

            if parsed.get("answer"):
                return str(parsed["answer"])

            if parsed.get("summary"):
                return str(parsed["summary"])

    except (json.JSONDecodeError, TypeError):
        pass

    return response

def generate_chat_response(question):

    method = st.session_state.chat_method

    if method == "Base Qwen":
        response = base_qwen.generate_response(question)

    elif method == "Fine-Tuned Qwen":
        response = finetuned_qwen.generate_response(question)

    elif method == "RAG":
        rag_result = rag.generate_response(question)
        if isinstance(rag_result, dict):
            response = rag_result.get("answer", "")
        else:
            response = rag_result

    else:
        response = "Please select an analysis method."

    return normalize_ai_response(response)


def render_chat():

    sparkles_svg = utils.load_svg("sparkles.svg")
    user_svg = utils.load_svg("user-round.svg")
    sparkles_img = utils.load_svg_as_image("sparkles.svg")

    with st.container(border=True):

        st.markdown(
            f"""
            <div class="chat-header">
                <div class="chat-title">
                    AI CHAT
                </div>
                <div class="chat-badge">
                    <span class="result-icon">
                        {sparkles_svg}
                    </span>
                    AI assistant
                </div>
            </div>
            """,
            unsafe_allow_html=True
        )

        st.session_state.chat_method = st.selectbox(
            "Analysis Method",
            [
                "Base Qwen",
                "Fine-Tuned Qwen",
                "RAG"
            ],
            index=[
                "Base Qwen",
                "Fine-Tuned Qwen",
                "RAG"
            ].index(st.session_state.chat_method),
            key="chat_method_selector"
        )

        for chat in st.session_state.messages:
            if chat["role"] == "user":
                st.markdown(
                    f"""
                      <div class="user-message">
                          <div class="user-bubble">
                              {chat["message"]}
                          </div>
                          <div class="user-icon">
                              {user_svg}
                          </div>
                      </div>
                    """,
                    unsafe_allow_html=True
                )

            else:
                formatted_message = markdown.markdown(chat["message"])

                st.html(
                    f"""
                    <div class="chat-message">
                        <div class="chat-icon">
                            <img src="{sparkles_img}" alt="AI">
                        </div>
                        <div class="chat-bubble">
                            {formatted_message}
                        </div>
                    </div>
                    """
                )

        with st.form("chat_form", clear_on_submit=True):

            chat_col1, chat_col2 = st.columns([10, 1])

            with chat_col1:

                user_question = st.text_input(
                    "Ask a question",
                    placeholder="Ask about companies, risk, sectors...",
                    label_visibility="collapsed"
                )

            with chat_col2:

                send = st.form_submit_button("➤")

        if send and user_question.strip():

            question = user_question.strip()

            st.session_state.messages.append({
                "role": "user",
                "message": question
            })

            response = generate_chat_response(question)

            st.session_state.messages.append({
                "role": "assistant",
                "message": response
            })

            st.rerun()
```

must always display human-readable text.

2. Extract the answer/summary field.

4. Return only plain text to the UI.

RAG source metadata remains separate from the displayed answer.

## Add GPU-Safe Model Switching

Colab and Streamlit share the same Python process, so previously loaded
Qwen models can remain in GPU memory.

The chat layer will release the unused Qwen model before loading the
selected method.

Only one Qwen model will be kept on the GPU at a time.

```python
from pathlib import Path

chat_path = Path("/content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/chat.py")

text = chat_path.read_text(encoding="utf-8")

# Add GPU cleanup imports
if "import gc" not in text:
    text = text.replace(
        "import json\nimport streamlit as st",
        "import gc\nimport json\nimport torch\nimport streamlit as st"
    )

# Add model-release helper before generate_chat_response()
marker = "def generate_chat_response(question):"

helper = '''def release_qwen_models():

    if getattr(base_qwen, "_model", None) is not None:
        del base_qwen._model
        base_qwen._model = None

    if getattr(base_qwen, "_tokenizer", None) is not None:
        base_qwen._tokenizer = None

    if getattr(finetuned_qwen, "_model", None) is not None:
        del finetuned_qwen._model
        finetuned_qwen._model = None

    if getattr(finetuned_qwen, "_tokenizer", None) is not None:
        finetuned_qwen._tokenizer = None

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()


'''

if "def release_qwen_models(" not in text:
    text = text.replace(marker, helper + marker)

# Replace generate_chat_response()
start = text.index("def generate_chat_response(")
end = text.index("\ndef render_chat():", start)

new_function = '''def generate_chat_response(question):

    method = st.session_state.chat_method

    if method == "Base Qwen":

        if getattr(finetuned_qwen, "_model", None) is not None:
            release_qwen_models()

        response = base_qwen.generate_response(question)

    elif method == "Fine-Tuned Qwen":

        if getattr(base_qwen, "_model", None) is not None:
            release_qwen_models()

        response = finetuned_qwen.generate_response(question)

    elif method == "RAG":

        if getattr(base_qwen, "_model", None) is not None:
            release_qwen_models()

        rag_result = rag.generate_response(question)
        response = rag_result["answer"]

    else:
        response = "Please select an analysis method."

    return normalize_ai_response(response)
'''

text = text[:start] + new_function + text[end:]

chat_path.write_text(text, encoding="utf-8")

print("chat.py updated with GPU-safe model switching.")
```

### Locate the Remaining Qwen Model

The Qwen module references were cleared, but another notebook variable
still holds Qwen model parameters on the GPU.

We will inspect the notebook's global variables and identify which
model object is still using CUDA memory.

```python
print("Global model-like objects currently using CUDA:\n")

for name, obj in list(globals().items()):
    try:
        if isinstance(obj, torch.nn.Module):
            cuda_params = [
                p for p in obj.parameters()
                if p.is_cuda
            ]

            if cuda_params:
                total_mb = sum(
                    p.numel() * p.element_size()
                    for p in cuda_params
                ) / 1024**2

                print(
                    f"{name}: "
                    f"{type(obj).__name__} | "
                    f"{total_mb:.2f} MB on GPU"
                )
    except Exception:
        pass
```

```python
# Find Indirect References to the GPU Model
print("Potential objects holding a model:\n")

for name, obj in list(globals().items()):
    try:
        if (
            hasattr(obj, "model")
            or hasattr(obj, "module")
            or hasattr(obj, "base_model")
        ):
            print(
                f"{name}: {type(obj).__name__}"
            )
    except Exception:
        pass
```

```python
# Identify the Remaining Qwen Model Reference
print("Searching for large CUDA tensors...\n")

found = []

for obj in gc.get_objects():
    try:
        if isinstance(obj, torch.Tensor) and obj.is_cuda:
            size_mb = obj.numel() * obj.element_size() / 1024**2

            if size_mb > 20:
                found.append((size_mb, tuple(obj.shape), type(obj).__name__))

    except Exception:
        pass

found.sort(reverse=True)

for size_mb, shape, obj_type in found[:30]:
    print(
        f"{size_mb:8.2f} MB | "
        f"shape={shape} | "
        f"type={obj_type}"
    )
```

```python
# Trace the Remaining Qwen Model
largest_qwen_parameter = None
largest_size_mb = 0

for obj in gc.get_objects():
    try:
        if isinstance(obj, torch.nn.Parameter) and obj.is_cuda:
            size_mb = obj.numel() * obj.element_size() / 1024**2

            if size_mb > largest_size_mb:
                largest_size_mb = size_mb
                largest_qwen_parameter = obj

    except Exception:
        pass

print(f"Largest CUDA Parameter: {largest_size_mb:.2f} MB")
print(f"Shape: {tuple(largest_qwen_parameter.shape)}")
print("\nObjects directly referencing this parameter:\n")

for ref in gc.get_referrers(largest_qwen_parameter):
    try:
        if isinstance(ref, dict):
            print(
                "DICT | keys:",
                list(ref.keys())[:10]
            )
        else:
            print(
                "OBJECT | type:",
                type(ref).__name__
            )
    except Exception:
        pass
```

```python
# Find the large Qwen parameter again
qwen_param = None

for obj in gc.get_objects():
    try:
        if (
            isinstance(obj, torch.nn.Parameter)
            and obj.is_cuda
            and tuple(obj.shape) == (151936, 1536)
        ):
            qwen_param = obj
            break
    except Exception:
        pass

print("Qwen parameter found:", qwen_param is not None)

if qwen_param is not None:
    print("\nDirect referrer types:\n")

    seen = set()

    for ref in gc.get_referrers(qwen_param):
        try:
            ref_type = type(ref).__name__

            if ref_type in seen:
                continue

            seen.add(ref_type)

            print(
                f"- {ref_type}"
            )

            if isinstance(ref, dict):
                print(
                    "  keys:",
                    list(ref.keys())[:20]
                )
            else:
                print(
                    "  repr:",
                    repr(ref)[:300]
                )

        except Exception:
            pass
```

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/dashboard.py

import streamlit as st
import pandas as pd

from modules import utils
from modules import investment_score
from modules import styles
from modules import classical_ml

# Display the InvestIQ landing page before an analysis is performed
def render_home():
  brain_svg = utils.load_svg("brain-circuit.svg")
  chart_svg = utils.load_svg("chart-column.svg")
  shield_svg = utils.load_svg("shield-check.svg")
  sparkles_svg = utils.load_svg("sparkles.svg")
  chat_svg = utils.load_svg("message-square.svg")

  st.html(
    f"""
      <div class="home-page">
        <div class="home-logo">
          <div class="home-logo-icon">
            {brain_svg}
          </div>

            <div class="home-logo-text">
              <div class="home-title">
                InvestIQ — AI Investment Research
              </div>
              <div class="home-subtitle">
                AI Investment Research Assistant
              </div>
            </div>
          </div>
          <div class="description-title">
              Analyze investments with an
            <span class="ai-gradient">AI analyst</span>
          </div>
          <div class="description">
            Describe your investment preferences in plain language.
            InvestIQ analyzes companies, identifies the strongest matches,
            and explains the reasoning behind each result.
          </div>
        </div>
    """
  )

  f1, f2, f3, f4 = st.columns(4)

  with f1:
    st.markdown(
      f"""
        <div class="feature">
          <span class="feature-icon">{chat_svg}</span>
            Conversational AI
        </div>
      """,
      unsafe_allow_html=True
      )

  with f2:
    st.markdown(
        f"""
          <div class="feature">
            <span class="feature-icon">{chart_svg}</span>
              Smart Ranking
          </div>
        """,
        unsafe_allow_html=True
        )

  with f3:
    st.markdown(
        f"""
          <div class="feature">
            <span class="feature-icon">{shield_svg}</span>
              Transparent Scoring
          </div>
        """,
        unsafe_allow_html=True
       )

  with f4:
    st.markdown(
        f"""
          <div class="feature">
            <span class="feature-icon">{sparkles_svg}</span>
              Visual Analytics
          </div>
        """,
        unsafe_allow_html=True
        )

  st.markdown("""<div class="preferences-separator"></div>""", unsafe_allow_html=True)

def render_navigation(best_company, companies_scanned):
  nav1, nav2, nav3 = st.columns([4, 3, 3])

  with nav1:
    st.markdown(
        f"""
          <div class="status-badge">
            <span class="status-dot"></span>
              Top Ranked: {best_company["Ticker"]}
          </div>
        """,
        unsafe_allow_html=True
       )

  with nav2:
    risk_level = investment_score.get_risk_level(best_company["Historical Volatility"])

    st.markdown(
      f"""
        <div class="status-badge">
          <span class="status-dot"></span>
            {best_company["Ticker"]} Risk:
            &nbsp; {risk_level}
        </div>
      """,
      unsafe_allow_html=True
      )

  with nav3:
    st.markdown(
      f"""
        <div class="live-badge">
            ● &nbsp; {companies_scanned} companies screened
        </div>
      """,
      unsafe_allow_html=True
      )

# Display key performance indicators for the selected Top 5 companies
def render_kpis(top_5, best_company):
  """
    Identifies the company with the highest revenue growth, highest
    profit margin, lowest risk, highest historical return, and the
    top-ranked company.
    """
  highest_growth = top_5.loc[top_5["Revenue Growth"].idxmax()]

  highest_profit = top_5.loc[top_5["Net Profit Margin"].idxmax()]

  lowest_risk = top_5.loc[top_5["Historical Volatility"].idxmin()]

  highest_return = top_5.loc[top_5["Historical Return"].idxmax()]

  c1, c2, c3, c4, c5 = st.columns(5)

  with c1:
    st.markdown(
      f"""
        <div class="kpi-card">
          <div class="kpi-label">
                Highest Revenue Growth
          </div>
          <div class="kpi-value" style="color:#00D6FF;">
              {highest_growth["Revenue Growth"] * 100:.1f}%
          </div>
          <div class="kpi-description">
              {highest_growth["Company_Name"]}
          </div>
        </div>
      """,
      unsafe_allow_html=True
      )

  with c2:
    st.markdown(
      f"""
        <div class="kpi-card">
          <div class="kpi-label">
              Highest Profit Margin
          </div>
          <div class="kpi-value" style="color:#00FF88;">
              {highest_profit["Net Profit Margin"] * 100:.1f}%
          </div>
          <div class="kpi-description">
              {highest_profit["Company_Name"]}
          </div>
        </div>
      """,
      unsafe_allow_html=True
      )

  with c3:
    st.markdown(
      f"""
        <div class="kpi-card">
          <div class="kpi-label">
              Lowest Risk
          </div>
          <div class="kpi-value" style="color:#C96BFF;">
              {investment_score.get_risk_level(lowest_risk["Historical Volatility"])}
          </div>
          <div
              class="kpi-description"
              style="color:#A05ACB;"
          >
              {lowest_risk["Company_Name"]}
          </div>
        </div>
      """,
      unsafe_allow_html=True
      )

  with c4:
    st.markdown(
      f"""
        <div class="kpi-card">
          <div class="kpi-label">
              Highest Historical Return
          </div>
          <div class="kpi-value" style="color:#00FF88;">
              {highest_return["Historical Return"] * 100:.1f}%
          </div>
          <div class="kpi-description">
              {highest_return["Company_Name"]}
          </div>
        </div>
      """,
      unsafe_allow_html=True
      )

  with c5:
    st.markdown(
      f"""
        <div class="kpi-card">
          <div class="kpi-label">
              Top Ranked Company
          </div>
          <div class="kpi-value" style="color:#00FF88;">
              {best_company["Company_Name"]}
          </div>
          <div class="kpi-description">
              {best_company["Investment Score"]:.0f}/100
          </div>
        </div>
      """,
      unsafe_allow_html=True
      )

# Display the Top 5 companies and their main financial and ML metrics
def render_company_table(top_5):
  """
    Includes growth, profit margin, risk level, Investment Score, and
    the estimated probability of a positive 1-year return.
  """
  companies = top_5[
      [
          "Rank",
          "Company_Name",
          "Ticker",
          "Sector",
          "Revenue Growth",
          "Net Profit Margin",
          "Historical Volatility",
          "Investment Score",
          "Positive Return Probability"
      ]
  ].copy()

  companies = companies.rename(
      columns={
          "Company_Name":"Company",
          "Net Profit Margin":"Profit Margin",
          "Investment Score":"Score",
          "Positive Return Probability": "Positive 1-Year Return Probability"
      }
  )

  companies["Risk"] = companies["Historical Volatility"].apply(investment_score.get_risk_level)

  companies["Revenue Growth"] = (companies["Revenue Growth"]*100).round(1)

  companies["Profit Margin"] = (companies["Profit Margin"]*100).round(1)

  companies["Positive 1-Year Return Probability"] = (companies["Positive 1-Year Return Probability"] * 100).round(1)

  companies["Score"] = (companies["Score"].round().astype(int))

  companies = companies[
      [
          "Rank",
          "Company",
          "Ticker",
          "Sector",
          "Revenue Growth",
          "Profit Margin",
          "Risk",
          "Score",
          "Positive 1-Year Return Probability"
      ]
  ]

  styled_companies = (companies.style
      .map(styles.style_company, subset=["Company"])
      .map(styles.style_ticker, subset=["Ticker"])
      .map(styles.style_percentage, subset=["Revenue Growth", "Profit Margin", "Positive 1-Year Return Probability"])
      .map(styles.style_risk, subset=["Risk"])
      .map(styles.style_score, subset=["Score"])
      .format({"Revenue Growth": "{:.1f}%", "Profit Margin": "{:.1f}%", "Score": "{:.0f}", "Positive 1-Year Return Probability": "{:.1f}%"})
      )

  with st.container(border=True):
    st.subheader("Top Companies")

    st.caption(
      "Estimated probability of a positive 1-year return. "
      "This is a model estimate, not a guarantee of future performance."
    )

    st.dataframe(
      styled_companies,
      use_container_width=True,
      hide_index=True
    )

# Run the dashboard analysis using the user's selected preferences
def render_dashboard():
  """
    Calculates the Top 5 companies, obtains the classical ML probability
    for each company, stores the results in session state, and renders
    the dashboard components.
  """
  settings = st.session_state.analysis_settings

  top_5, companies_scanned = investment_score.get_top_5(
      sector=settings["sector"],
      risk=settings["risk"],
      goal=settings["goal"]
  )

  # Add classical ML prediction to the selected Top 5
  for index in top_5.index:
    company_row = top_5.loc[[index]]

    probability = classical_ml.predict_positive_return_probability(company_row)

    top_5.loc[index, "Positive Return Probability"] = probability

  best_company = top_5.iloc[0]

  st.session_state.top_5 = top_5
  st.session_state.best_company = best_company

  render_navigation(best_company, companies_scanned)

  render_kpis(top_5, best_company)

  render_company_table(top_5)
```

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/investment_score.py

import pandas as pd

dataset = "/content/drive/MyDrive/data_handling_and_machine_learning/datasets/merged_dataset/merged_dataset.csv"
company_info_file = "/content/drive/MyDrive/data_handling_and_machine_learning/week_03/outputs/olga_companies_info.csv"

def normalize_score(series):
  """
    Convert values to percentile-based scores from 0 to 100.

    Missing values receive a neutral score of 50.
  """
  scores = series.rank(pct=True, method="average") * 100

  return scores.fillna(50)

def calculate_profitability(df):
  net_margin = normalize_score(df["Net Profit Margin"])

  operating_margin = normalize_score(df["Operating Profit Margin"])

  fcf_margin = normalize_score(df["FCF Margin"])

  raw_score = (net_margin * 0.40 + operating_margin * 0.30 + fcf_margin * 0.30)

  df["Profitability Score"] = raw_score * 0.30

  return df

def calculate_historical_growth(df):
  revenue_growth = normalize_score(df["Revenue Growth"])

  net_income_growth = normalize_score(df["Net Income Growth"])

  operating_income_growth = normalize_score(df["Operating Income Growth"])

  asset_growth = normalize_score(df["Asset Growth"])

  raw_score = (revenue_growth * 0.40 + net_income_growth * 0.30 + operating_income_growth * 0.20 + asset_growth * 0.10)

  df["Historical Growth Score"] = raw_score * 0.25

  return df

def calculate_financial_stability(df):
  current_ratio = normalize_score(df["Current Ratio"])

  debt = 100 - normalize_score(df["Debt-to-Equity"])

  free_cash_flow = normalize_score(df["Free Cash Flow"])

  equity = normalize_score(df["Total Equity"])

  raw_score = (current_ratio * 0.25 + debt * 0.30 + free_cash_flow * 0.25 + equity * 0.20)

  df["Financial Stability Score"] = raw_score * 0.25

  return df

def calculate_historical_return(df):
  historical_return = normalize_score(df["Historical Return"])

  low_volatility = (100 - normalize_score(df["Historical Volatility"]))

  raw_score = (historical_return * 0.70 + low_volatility * 0.30)

  df["Historical Return Score"] = raw_score * 0.20

  return df

def calculate_investment_score(df):
  df = df.copy()

  df = calculate_profitability(df)

  df = calculate_historical_growth(df)

  df = calculate_financial_stability(df)

  df = calculate_historical_return(df)

  df["Investment Score"] = (
      df["Profitability Score"]
      + df["Historical Growth Score"]
      + df["Financial Stability Score"]
      + df["Historical Return Score"]
  )

  return df

def get_risk_level(volatility):
  if pd.isna(volatility):
      return "Medium"
  if volatility <= 0.20:
      return "Low"
  elif volatility <= 0.35:
      return "Medium"
  else:
      return "High"

def load_company_data():
  financials = pd.read_csv(dataset)

  company_info = pd.read_csv(company_info_file)

  df = financials.merge(
      company_info[["Ticker", "SimFinId", "Company_Name", "Sector", "Industry", "Currency"]],
      on="Ticker",
      how="left"
  )

  return df

def get_latest_companies(df):
  df = df.sort_values("Fiscal Year")

  latest = (df.groupby("Ticker",as_index=False)
      .tail(1)
      .copy()
  )

  return latest

def apply_preferences(df, sector, risk, goal):
  filtered = df.copy()

  #sector
  if sector != "All":
    filtered = filtered[filtered["Sector"] == sector].copy()

  # risk
  volatility_percentile = normalize_score(filtered["Historical Volatility"])

  if risk == "Low":
    risk_match = 100 - volatility_percentile

  elif risk in ["Moderate", "Medium"]:
    percentile = (filtered["Historical Volatility"].rank(pct=True))

    risk_match = (1 - abs(percentile - 0.5) * 2) * 100

    risk_match = risk_match.fillna(50)

  elif risk == "High":
    risk_match = volatility_percentile

  else:
    risk_match = pd.Series(50, index=filtered.index)

  #goal
  if goal == "High Profitability":
    goal_match = (
        normalize_score(filtered["Net Profit Margin"]) * 0.40
        +
        normalize_score(filtered["Operating Profit Margin"]) * 0.30
        +
        normalize_score(filtered["FCF Margin"]) * 0.30
    )

  elif goal == "Maximum Growth":
    goal_match = (
        normalize_score(filtered["Revenue Growth"]) * 0.40
        +
        normalize_score(filtered["Net Income Growth"]) * 0.30
        +
        normalize_score(filtered["Operating Income Growth"]) * 0.20
        +
        normalize_score(filtered["Asset Growth"]) * 0.10
    )

  elif goal == "Long Term Stability":
    goal_match = (filtered["Financial Stability Score"] / 25 * 100)

  elif goal == "Fast Investment Recovery":
    goal_match = normalize_score(filtered["Historical Return"])

  elif goal == "Balanced Return and Risk":
    return_match = normalize_score(filtered["Historical Return"])

    low_volatility_match = (100 - normalize_score(filtered["Historical Volatility"]))

    goal_match = (return_match * 0.50 + low_volatility_match * 0.50)

  else:
    goal_match = pd.Series(50, index=filtered.index)

  filtered["Goal Match"] = goal_match
  filtered["Risk Match"] = risk_match

  filtered["Preference Score"] = (
      filtered["Goal Match"] * 0.70
      + filtered["Risk Match"] * 0.30
  )

  filtered["Ranking Score"] = (
      filtered["Investment Score"]
      + filtered["Preference Score"] * 0.25
  )

  return filtered

def get_top_5(sector, risk, goal):
  df = load_company_data()

  latest = get_latest_companies(df)

  latest = calculate_investment_score(latest)

  filtered = apply_preferences(latest, sector, risk, goal)

  companies_scanned = filtered["Ticker"].nunique()

  top_5 = (
      filtered
      .sort_values("Ranking Score", ascending=False)
      .head(5)
      .reset_index(drop=True)
  )

  top_5["Rank"] = (top_5.index + 1)

  return top_5, companies_scanned
```

# Classical ML Prediction Module

This module loads the trained Decision Tree, the saved 31-feature list, and the median imputer used during training.

Its only responsibility is to generate the probability that a company's 1-year future return will be positive.

The existing investment-ranking system, fine-tuned Qwen model, base Qwen model, and RAG system remain separate components.

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/classical_ml.py

import json
import joblib
import pandas as pd


MODEL_PATH = ("/content/drive/MyDrive/data_handling_and_machine_learning/datasets/classical_ml_classification_dataset/InvestIQ_decision_tree_model.pkl")

FEATURES_PATH = ("/content/drive/MyDrive/data_handling_and_machine_learning/datasets/classical_ml_classification_dataset/InvestIQ_classical_ml_features.json")

IMPUTER_PATH = ("/content/drive/MyDrive/data_handling_and_machine_learning/datasets/classical_ml_classification_dataset/InvestIQ_classical_ml_imputer.pkl")

_model = None
_imputer = None
_features = None

def load_artifacts():
    global _model, _imputer, _features

    if _model is None:
        _model = joblib.load(MODEL_PATH)

    if _imputer is None:
        _imputer = joblib.load(IMPUTER_PATH)

    if _features is None:
        with open(FEATURES_PATH, "r", encoding="utf-8") as f:
            _features = json.load(f)

    return _model, _imputer, _features

def predict_positive_return_probability(company_data):
    """
    Predict the probability of a positive 1-year future return.

    company_data must contain the 31 features used during training.
    """

    model, imputer, features = load_artifacts()

    input_data = company_data[features].copy()

    input_imputed = imputer.transform(input_data)

    input_imputed = pd.DataFrame(
        input_imputed,
        columns=features,
        index=input_data.index
    )

    probability = model.predict_proba(input_imputed)[0][1]

    return float(probability)
```

# Base Qwen Streamlit Connection

Create a small reusable module for Base Qwen.

The module will load the Qwen model once and provide a function that accepts a user question and returns the generated response.

```python
from pathlib import Path

base_qwen_path = Path("/content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/base_qwen.py")

base_qwen_code = r'''
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"

_tokenizer = None
_model = None

def load_model():
    global _tokenizer, _model

    if _tokenizer is None:
        _tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
        _tokenizer.pad_token = _tokenizer.eos_token

    if _model is None:
        _model = AutoModelForCausalLM.from_pretrained(
            MODEL_NAME,
            device_map="cuda",
            torch_dtype=torch.float16
        )

        _model.config.use_cache = True
        _model.eval()

    return _tokenizer, _model

def generate_response(question):
    tokenizer, model = load_model()

    messages = [
        {
            "role": "system",
            "content": (
                "You are InvestIQ, an AI investment research assistant. "
                "Provide clear, structured financial analysis and do not "
                "invent information that is not available."
            )
        },
        {
            "role": "user",
            "content": question
        }
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=300,
            do_sample=False
        )

    input_length = inputs["input_ids"].shape[-1]

    response = tokenizer.decode(
        outputs[0][input_length:],
        skip_special_tokens=True
    )

    return response
'''

base_qwen_path.write_text(
    base_qwen_code,
    encoding="utf-8"
)

print("Created:")
print(base_qwen_path)
```

# Connect the Existing RAG Pipeline to Streamlit

The existing RAG retrieval and reranking logic is preserved.

The Streamlit RAG module will:
1. Embed the user's question with MiniLM.
2. Retrieve candidates from the existing Chroma collection.
3. Rerank them with the existing Cross-Encoder.
4. Build the existing grounded RAG prompt.
5. Use the existing Fine-Tuned Qwen module for generation.

The Fine-Tuned Qwen model is not implemented or loaded separately here.

```python
rag_module_path = (
    "/content/drive/MyDrive/"
    "data_handling_and_machine_learning/"
    "UI/modules/rag.py"
)

rag_code = r'''
import chromadb
from sentence_transformers import SentenceTransformer, CrossEncoder

import finetuned_qwen


CHROMA_PATH = (
    "/content/drive/MyDrive/"
    "data_handling_and_machine_learning/"
    "chroma_db"
)

COLLECTION_NAME = "investiq_rag"

_embedding_model = None
_reranker = None
_collection = None


def load_rag_components():
    global _embedding_model
    global _reranker
    global _collection

    if _embedding_model is None:
        _embedding_model = SentenceTransformer(
            "all-MiniLM-L6-v2"
        )

    if _reranker is None:
        _reranker = CrossEncoder(
            "cross-encoder/ms-marco-MiniLM-L-6-v2"
        )

    if _collection is None:
        chroma_client = chromadb.PersistentClient(
            path=CHROMA_PATH
        )

        _collection = chroma_client.get_collection(
            name=COLLECTION_NAME
        )

    return _embedding_model, _reranker, _collection


def retrieve_chunks(
    query,
    n_candidates=10,
    n_results=5,
    company=None
):
    embedding_model, reranker, collection = load_rag_components()

    query_embedding = embedding_model.encode(
        query
    ).tolist()

    if company is not None:
        results = collection.query(
            query_embeddings=[query_embedding],
            n_results=n_candidates,
            where={"company": company}
        )
    else:
        results = collection.query(
            query_embeddings=[query_embedding],
            n_results=n_candidates
        )

    candidates = []

    for document, metadata, distance in zip(
        results["documents"][0],
        results["metadatas"][0],
        results["distances"][0]
    ):
        candidates.append({
            "text": document,
            "metadata": metadata,
            "distance": distance
        })

    candidate_pairs = [
        (query, result["text"])
        for result in candidates
    ]

    scores = reranker.predict(candidate_pairs)

    reranked = []

    for result, score in zip(candidates, scores):
        reranked.append({
            "text": result["text"],
            "metadata": result["metadata"],
            "distance": result["distance"],
            "reranker_score": float(score)
        })

    reranked.sort(
        key=lambda x: x["reranker_score"],
        reverse=True
    )

    return reranked[:n_results]


def build_reranked_context(retrieved_chunks):
    context_parts = []

    for rank, result in enumerate(
        retrieved_chunks,
        start=1
    ):
        metadata = result["metadata"]

        context_parts.append(
            f"""
Source ID: {metadata["chunk_id"]}
Company: {metadata["company"]}
Section: {metadata["section"]}
Retrieval Rank: {rank}

Text:
{result["text"]}
"""
        )

    return "\n".join(context_parts)


def generate_response(question, company=None):
    reranked_results = retrieve_chunks(
        query=question,
        n_candidates=10,
        n_results=5,
        company=company
    )

    reranked_context = build_reranked_context(
        reranked_results
    )

    rag_prompt_v2 = f"""
You are an investment research assistant.

Answer the user's question using ONLY the retrieved source context below.

GROUNDING RULES:
1. Treat the retrieved context as the complete evidence boundary.
2. Use only information explicitly supported by the retrieved context.
3. Do not use outside knowledge to fill missing information.
4. If the context does not contain enough evidence to answer the question,
   explicitly state that the available sources do not provide enough information.
5. Cite a source ID only when the answer uses a factual detail supported by that source.
6. Do not cite a source merely because it was included in the retrieved context.
7. Do not cite irrelevant sources.
8. Never invent source IDs.
9. Every cited source ID must appear in the retrieved context.
10. Keep the answer concise and directly related to the user's question.

Return ONLY valid JSON using exactly this structure:

{{
  "answer": "A concise grounded answer.",
  "sources": ["source_id_1", "source_id_2"]
}}

User question:
{question}

Retrieved context:
{reranked_context}
"""

    return finetuned_qwen.generate_response(
        rag_prompt_v2
    )
'''

with open(rag_module_path, "w", encoding="utf-8") as f:
    f.write(rag_code)

print("rag.py created successfully.")
print(rag_module_path)
```
