# Run app

```python
import subprocess
import time
import os

# Ensure output files exist and are empty
with open("streamlit_stdout.log", "w") as f:
    pass
with open("streamlit_stderr.log", "w") as f:
    pass

subprocess.run(
    ["pkill", "-f", "streamlit"],
    stderr=subprocess.DEVNULL,
    stdout=subprocess.DEVNULL
)

# Open files to capture stdout and stderr
stdout_file = open("streamlit_stdout.log", "a")
stderr_file = open("streamlit_stderr.log", "a")

# 3. Launch Streamlit, capturing its output
streamlit_process = subprocess.Popen(
    [
        "streamlit", "run", "app.py",
        "--server.port", "8501",
        "--server.headless", "true",
        "--server.enableCORS", "false",
        "--server.enableXsrfProtection", "false"
    ],
    stdout=stdout_file,
    stderr=stderr_file,
    preexec_fn=os.setsid # To ensure the process group is killed later
)

# 4. Give Streamlit time to start
time.sleep(5) # Increased sleep time for more robust startup

# Close the files after sleep to ensure all output is flushed
stdout_file.close()
stderr_file.close()

# Print the captured logs
print("\n--- Streamlit Server STDOUT ---")
with open("streamlit_stdout.log", "r") as f:
    print(f.read())
print("\n--- Streamlit Server STDERR ---")
with open("streamlit_stderr.log", "r") as f:
    print(f.read())

# 5. Get Colab's public proxy URL
from google.colab import output
from google.colab.output import eval_js

proxy_url = eval_js("google.colab.kernel.proxyPort(8501)")

print("=" * 70)
print("INVESTIQ IS RUNNING")
print("=" * 70)
print(f"Web address: {proxy_url}")
print("=" * 70)

# 6. Display the app directly inside Colab
output.serve_kernel_port_as_iframe(8501, height="800")
```

# Chat

```python
%%writefile chat.py

import streamlit as st
import utils

def initialize_chat():
    if "messages" not in st.session_state:
        st.session_state.messages = [
            {
                "role": "assistant",
                "message": (
                    "Hi! I'm your INVESTIQ research assistant. "
                    "Ask me anything about the analyzed companies, "
                    "risk, sectors, or your investment strategy."
                )
            }
        ]

def render_chat():

    sparkles_svg = utils.load_svg("sparkles.svg")
    user_svg = utils.load_svg("user-round.svg")

    with st.container(border=True):
        st.markdown(
            f"""
            <div class="chat-header">
                <div class="chat-title">
                    AI CHAT
                </div>
                <div class="chat-badge">
                    <span class="result-icon">
                        {sparkles_svg}
                    </span>
                    AI assistant
                </div>
            </div>
            """,
            unsafe_allow_html=True
        )

        chat_box = st.container(height=330, border=False)

        with chat_box:
            for chat in st.session_state.messages:
                if chat["role"] == "user":
                    st.markdown(
                        f"""
                        <div class="user-message">
                            <div class="user-bubble">
                                {chat["message"]}
                            </div>
                            <div class="user-icon">
                                {user_svg}
                            </div>
                        </div>
                        """, unsafe_allow_html=True
                    )

                else:
                    st.markdown(
                        f"""
                        <div class="chat-message">
                            <div class="chat-icon">
                                {sparkles_svg}
                            </div>
                            <div class="chat-bubble">
                                {chat["message"]}
                            </div>
                        </div>
                        """, unsafe_allow_html=True
                    )

        chat_col1, chat_col2 = st.columns([12, 1])

        with chat_col1:
            user_question = st.text_input(
                "Ask a question",
                placeholder="Ask about companies, risk, sectors...",
                label_visibility="collapsed"
            )

        with chat_col2:
            send = st.button(
                "➤",
                key="send_chat"
            )

        if send and user_question.strip():
            question = user_question.strip()

            st.session_state.messages.append({
                "role": "user",
                "message": question
            })

            mock_response = (
                "Based on the current investment analysis, "
                "this company has a moderate risk profile. "
                "Its historical performance suggests potential "
                "for growth, although investors should consider "
                "sector volatility and diversification."
            )

            st.session_state.messages.append({
                "role": "assistant",
                "message": mock_response
            })

            st.rerun()
```

# Analytics

```python
%%writefile analysis.py

import streamlit as st
import pandas as pd
import plotly.graph_objects as go
import random

import utils


def render_revenue_chart(investment, horizon):

    years = [
        f"Year {i}"
        for i in range(1, horizon + 1)
    ]

    company_a = [70, 82, 95, 88, 94][:horizon]
    company_b = [48, 57, 63, 65, 68][:horizon]
    company_c = [60, 56, 60, 65, 72][:horizon]
    company_d = [40, 23, 50, 47, 40][:horizon]
    company_e = [50, 52, 60, 61, 65][:horizon]

    fig = go.Figure()

    fig.add_trace(
        go.Scatter(
            x=years,
            y=company_a,
            mode="lines+markers",
            name="Company A",
            line=dict(color=utils.BLUE),
            marker=dict(color=utils.BLUE),
        )
    )

    fig.add_trace(
        go.Scatter(
            x=years,
            y=company_b,
            mode="lines+markers",
            name="Company B",
            line=dict(color=utils.BLUE_PURPLE),
            marker=dict(color=utils.BLUE_PURPLE),
        )
    )

    fig.add_trace(
        go.Scatter(
            x=years,
            y=company_c,
            mode="lines+markers",
            name="Company C",
            line=dict(color=utils.PURPLE),
            marker=dict(color=utils.PURPLE),
        )
    )

    fig.add_trace(
        go.Scatter(
            x=years,
            y=company_d,
            mode="lines+markers",
            name="Company D",
            line=dict(color=utils.GREEN),
            marker=dict(color=utils.GREEN),
        )
    )

    fig.add_trace(
        go.Scatter(
            x=years,
            y=company_e,
            mode="lines+markers",
            name="Company E",
            line=dict(color=utils.YELLOW),
            marker=dict(color=utils.YELLOW),
        )
    )

    fig.update_layout(

        template="plotly_dark",

        title=dict(
            text=(
                f"<b>{horizon}-Year Revenue Growth</b>"
                "<br>"
                f"<span style='font-size:12px; color:#64748B;'>"
                f"Projected returns by target market · "
                f"${investment/1_000_000:.1f}M capital deployed"
                "</span>"
            ),

            x=0.025,
            xanchor="left",
            y=0.95,
            yanchor="top",

            font=dict(
                size=14,
                color="#A0A8B9"
            )
        ),

        height=380,

        margin=dict(
            l=55,
            r=20,
            t=75,
            b=80
        ),

        yaxis=dict(
            range=[0, 120],
            tickvals=[0, 30, 60, 90, 120],
            gridcolor="#293241",
            zeroline=False,
            showline=False,

            tickfont=dict(
                color="#64748B",
                size=11
            )
        ),

        xaxis=dict(
            showgrid=False,
            zeroline=False,

            tickfont=dict(
                color="#64748B",
                size=12
            )
        ),

        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=-0.25,
            xanchor="left",
            x=0,

            font=dict(
                size=12
            ),

            bgcolor="rgba(0,0,0,0)"
        ),

        hovermode="x unified"
    )

    with st.container(border=True):

        st.plotly_chart(
            fig,
            use_container_width=True
        )


def render_investment_score(sector):

    left, right = st.columns([1, 1])

    bars = [
        ("Profitability", 25, 25, utils.GREEN),
        ("Historical Growth", 20, 14, utils.BLUE),
        ("Financial Stability", 20, 16, utils.BLUE_PURPLE),
        ("Future Growth", 20, 19, utils.PURPLE),
        ("Historical Return", 15, 12, utils.YELLOW),
    ]

    with left:

        with st.container(border=True):

            st.markdown(
                """
                <p style="
                    color:#FFFFFF;
                    font-size:14px;
                    font-weight:700;
                ">
                    INVESTMENT SCORE
                </p>
                """,
                unsafe_allow_html=True
            )

            st.html(
                f"""
                <div style="
                    display:flex;
                    justify-content:space-between;
                    align-items:center;
                    margin-bottom:25px;
                ">

                    <div style="
                        color:{utils.BLUE};
                        font-size:30px;
                        font-weight:700;
                    ">
                        COMPANY A
                    </div>

                    <div style="
                        color:#FFFFFF;
                        font-size:30px;
                        font-weight:700;
                    ">
                        84
                        <span style="
                            color:#526784;
                            font-size:24px;
                        ">
                            /100
                        </span>
                    </div>

                </div>
                """
            )

            for title, maximum, value, color in bars:

                percentage = (value / maximum) * 100

                st.html(
                    f"""
                    <div style="margin-bottom:12px;">

                        <div style="
                            display:flex;
                            justify-content:space-between;
                            color:#64748B;
                            font-size:14px;
                            margin-bottom:6px;
                        ">
                            <span>{title}</span>
                            <span>{value}/{maximum}</span>
                        </div>

                        <div style="
                            width:100%;
                            height:5px;
                            background:#202938;
                            border-radius:10px;
                        ">

                            <div style="
                                width:{percentage}%;
                                height:5px;
                                background:{color};
                                border-radius:10px;
                            "></div>

                        </div>

                    </div>
                    """
                )

    with right:

        with st.container(border=True):

            st.html(
                f"""
                <div class="recommendation-card">

                    <div class="recommendation-title">
                        AI RECOMMENDATION
                    </div>

                    <div class="recommendation-company">
                        COMPANY A
                    </div>

                    <div class="recommendation-sector">
                        {sector} · USA
                    </div>

                    <div class="recommendation-score">
                        Investment score: 84/100
                    </div>

                    <div class="why-title">
                        Why was it selected?
                    </div>

                    <div class="recommendation-text">
                        Company A ranks highest because it combines strong
                        profitability, consistent historical growth, a healthy
                        financial position and competitive projected growth.
                        Although another company may show a higher historical
                        return, Company A offers a stronger overall balance
                        between return and risk for your selected profile.
                    </div>

                </div>
                """
            )


def render_strengths_and_risks():

    c1, c2 = st.columns(2)

    with c1:

        with st.container(border=True):

            st.html(
                """
                <div class="strengths-card">

                    <div class="strengths-title">
                        STRENGTHS
                    </div>

                    <div class="strength-item">
                        Strong revenue growth
                    </div>

                    <div class="strength-item">
                        High profit margin
                    </div>

                    <div class="strength-item">
                        Consistent historical performance
                    </div>

                </div>
                """
            )

    with c2:

        with st.container(border=True):

            st.html(
                """
                <div class="risks-card">

                    <div class="risks-title">
                        RISKS
                    </div>

                    <div class="risk-item">
                        Market volatility
                    </div>

                    <div class="risk-item">
                        Industry competition
                    </div>

                    <div class="risk-item">
                        Future growth uncertainty
                    </div>

                </div>
                """
            )


def render_projection(investment, horizon):

    years = list(range(1, horizon + 1))

    values = []

    current_value = investment / 1_000_000

    for year in years:

        growth = random.uniform(0.05, 0.20)

        current_value *= (1 + growth)

        values.append(current_value)

    projection = pd.DataFrame({
        "Year": [
            f"Year {year}"
            for year in years
        ],

        "Value": values
    })

    with st.container(border=True):

        st.markdown(
            """
            <div style="
                color: #A0A8B9;
                font-size: 14px;
                font-weight: 600;
                margin-bottom: 14px;
            ">
                Investment Recovery Projection
            </div>
            """,
            unsafe_allow_html=True
        )

        col1, col2, col3, col4 = st.columns(4)

        card_style = """
        <div style="
            border: 1px solid #293241;
            border-radius: 6px;
            padding: 10px 12px;
            min-height: 60px;
            background-color: #11151D;
            box-sizing: border-box;
        ">

            <div style="
                color: #64748B;
                font-size: 11px;
                font-weight: 500;
                margin-bottom: 5px;
                line-height: 1.2;
            ">
                {title}
            </div>

            <div style="
                color: #FFFFFF;
                font-size: 17px;
                font-weight: 600;
                line-height: 1.2;
            ">
                {value}
            </div>

        </div>
        """

        initial_investment = investment / 1_000_000

        with col1:

            st.html(
                card_style.format(
                    title="INITIAL INVESTMENT",
                    value=f"${initial_investment:.2f}M"
                )
            )

        with col2:

            st.html(
                card_style.format(
                    title="ESTIMATED ANNUAL RETURN",
                    value="12.5%"
                )
            )

        with col3:

            st.html(
                card_style.format(
                    title="ESTIMATED RECOVERY PERIOD",
                    value="5.9 years"
                )
            )

        with col4:

            st.html(
                card_style.format(
                    title=f"VALUE IN ${horizon} YEARS",
                    value="$1.80M"
                )
            )

        st.markdown(
            "<div style='height: 20px;'></div>",
            unsafe_allow_html=True
        )

        fig = go.Figure()

        colors = [
            utils.BLUE,
            utils.BLUE_PURPLE,
            utils.PURPLE,
            utils.GREEN,
            utils.YELLOW
        ]

        max_value = max(values) * 1.15

        for i, row in projection.iterrows():

            color = colors[i % len(colors)]

            fig.add_trace(
                go.Bar(
                    x=[row["Year"]],
                    y=[row["Value"]],
                    name=row["Year"],

                    marker=dict(
                        color=color,
                        line=dict(width=0)
                    ),

                    text=[
                        f"{row['Value']:.2f}M"
                    ],

                    textposition="outside"
                )
            )

        fig.update_layout(

            height=300,

            margin=dict(
                l=55,
                r=20,
                t=20,
                b=60
            ),

            bargap=0.45,

            yaxis=dict(
                range=[0, max_value],
                gridcolor="#293241",
                gridwidth=1,
                zeroline=False,
                showline=False,

                tickfont=dict(
                    color="#64748B",
                    size=11
                )
            ),

            xaxis=dict(
                showgrid=False,
                zeroline=False,

                tickfont=dict(
                    color="#64748B",
                    size=12
                )
            ),

            legend=dict(
                orientation="h",
                yanchor="bottom",
                y=-0.32,
                xanchor="left",
                x=0,
                bgcolor="rgba(0,0,0,0)",

                font=dict(
                    size=12
                )
            ),

            showlegend=False
        )

        st.plotly_chart(
            fig,
            use_container_width=True,
            config={
                "displayModeBar": False
            }
        )


def render_analysis():

    settings = st.session_state.analysis_settings

    investment = settings["investment"]
    horizon = settings["horizon"]

    render_revenue_chart(
        investment,
        horizon
    )

    render_investment_score(settings["sector"])

    render_strengths_and_risks()

    render_projection(
        investment,
        horizon
    )
```
