# Analyzation (preferences.py)
---

**UI Layout Redesign**

The Investment Preferences section will remain below the AI Research chat.

The controls will be arranged horizontally where appropriate to reduce vertical space:

- Investment Amount remains full width.
- Sector and Risk share one row.
- Investment Goal gets its own row because its text is longer.
- Analyze Investments remains clearly separated as the main action.

The underlying session-state logic remains unchanged.

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/preferences.py

import streamlit as st

def render_preferences():
  st.markdown(
      """
      <div class="preferences-card">
          <div class="sidebar-title">
              Analyze Investments
          </div>
      </div>
      """,
      unsafe_allow_html=True
  )

  amount_column, goal_column, sector_column, risk_column = st.columns([1.2, 1.4, 1, 0.8])

  with amount_column:
      investment = st.number_input(
          "Investment Amount ($)",
          value=1_000_000
      )

  with goal_column:
      goal = st.selectbox(
          "Investment Goal",
          [
              "Balanced Return and Risk",
              "Long Term Stability",
              "Maximum Growth",
              "Fast Investment Recovery",
              "High Profitability"
          ]
      )

  with sector_column:
      sector = st.selectbox(
          "Investment Sector",
          [
              "All",
              "Technology",
              "Healthcare",
              "Energy",
              "Finance",
              "Consumer",
              "Industrials",
              "Communication",
              "Utilities",
              "Real Estate"
          ]
      )

  with risk_column:
      risk = st.selectbox(
          "Risk Tolerance",
          [
              "Low",
              "Moderate",
              "High"
          ]
      )

  empty_column, button_column = st.columns([3, 1])

  with button_column:
    analyze = st.button("Analyze Investments", use_container_width=True)

  if analyze:
      st.session_state.analyzed = True

      st.session_state.analysis_settings = {
          "investment": investment,
          "sector": sector,
          "risk": risk,
          "goal": goal
      }
```

# Chat

Fixes done:
- Update the chat input to use a Streamlit form so the question field clears automatically after sending.

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/chat.py

import streamlit as st
from modules import utils

def initialize_chat():
    if "messages" not in st.session_state:
        st.session_state.messages = [
            {
                "role": "assistant",
                "message": (
                    "Hi! I'm your INVESTIQ research assistant. "
                    "Ask me anything about the analyzed companies, "
                    "risk, sectors, or your investment strategy."
                )
            }
        ]


def render_chat():

    sparkles_svg = utils.load_svg("sparkles.svg")
    user_svg = utils.load_svg("user-round.svg")

    with st.container(border=True):

        st.markdown(
            f"""
            <div class="chat-header">
                <div class="chat-title">
                    AI CHAT
                </div>
                <div class="chat-badge">
                    <span class="result-icon">
                        {sparkles_svg}
                    </span>
                    AI assistant
                </div>
            </div>
            """,
            unsafe_allow_html=True
        )

        chat_box = st.container(height=330, border=False)

        with chat_box:
            for chat in st.session_state.messages:

                if chat["role"] == "user":

                    st.markdown(
                        f"""
                        <div class="user-message">
                            <div class="user-bubble">
                                {chat["message"]}
                            </div>
                            <div class="user-icon">
                                {user_svg}
                            </div>
                        </div>
                        """,
                        unsafe_allow_html=True
                    )

                else:

                    st.markdown(
                        f"""
                        <div class="chat-message">
                            <div class="chat-icon">
                                {sparkles_svg}
                            </div>
                            <div class="chat-bubble">
                                {chat["message"]}
                            </div>
                        </div>
                        """,
                        unsafe_allow_html=True
                    )

        with st.form("chat_form", clear_on_submit=True):
            chat_col1, chat_col2 = st.columns([10, 1])

            with chat_col1:
                user_question = st.text_input(
                    "Ask a question",
                    placeholder="Ask about companies, risk, sectors...",
                    label_visibility="collapsed"
                )

            with chat_col2:
                send = st.form_submit_button(
                    "➤"
                )

        if send and user_question.strip():
            question = user_question.strip()

            st.session_state.messages.append({
                "role": "user",
                "message": question
            })

            mock_response = (
                "Based on the current investment analysis, "
                "this company has a moderate risk profile. "
                "Its historical performance suggests potential "
                "for growth, although investors should consider "
                "sector volatility and diversification."
            )

            st.session_state.messages.append({
                "role": "assistant",
                "message": mock_response
            })

            st.rerun()
```

# Calculations

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/investment_score.py

import pandas as pd
import numpy as np

dataset = "/content/drive/MyDrive/data_handling_and_machine_learning/datasets/merged_dataset/merged_dataset.csv"
company_info_file = "/content/drive/MyDrive/data_handling_and_machine_learning/week_03/outputs/olga_companies_info.csv"

# missing values give a neutral score
def normalize_score(series):
    return series.rank(
        pct=True,
        method="average"
    ) * 100

def calculate_profitability(df):

    net_margin = normalize_score(df["Net Profit Margin"])
    operating_margin = normalize_score(df["Operating Profit Margin"])
    fcf_margin = normalize_score(df["FCF Margin"])
    raw_score = (
        net_margin * 0.40
        + operating_margin * 0.30
        + fcf_margin * 0.30
    )
    df["Profitability Score"] = (raw_score * 0.30)

    return df

def calculate_historical_growth(df):

    revenue_growth = normalize_score(df["Revenue Growth"])
    net_income_growth = normalize_score(df["Net Income Growth"])
    operating_income_growth = normalize_score(df["Operating Income Growth"])
    asset_growth = normalize_score(df["Asset Growth"])
    raw_score = (
        revenue_growth * 0.40
        + net_income_growth * 0.30
        + operating_income_growth * 0.20
        + asset_growth * 0.10
    )
    df["Historical Growth Score"] = (raw_score * 0.25)

    return df

def calculate_financial_stability(df):

    current_ratio = normalize_score(df["Current Ratio"])
    debt = 100 - normalize_score(df["Debt-to-Equity"])
    free_cash_flow = normalize_score(df["Free Cash Flow"])
    equity = normalize_score(df["Total Equity"])
    raw_score = (
        current_ratio * 0.25
        + debt * 0.30
        + free_cash_flow * 0.25
        + equity * 0.20
    )
    df["Financial Stability Score"] = (
        raw_score * 0.25
    )

    return df

def calculate_historical_return(df):

    historical_return = normalize_score(df["Historical Return"])
    low_volatility = (100 - normalize_score(df["Historical Volatility"]))
    raw_score = (historical_return * 0.70 + low_volatility * 0.30)
    df["Historical Return Score"] = (raw_score * 0.20)

    return df

def calculate_investment_score(df):

    df = df.copy()
    df = calculate_profitability(df)
    df = calculate_historical_growth(df)
    df = calculate_financial_stability(df)
    df = calculate_historical_return(df)

    df["Investment Score"] = (
        df["Profitability Score"]
        + df["Historical Growth Score"]
        + df["Financial Stability Score"]
        + df["Historical Return Score"]
    )

    return df

def get_risk_level(volatility):

    if pd.isna(volatility):
        return "Medium"
    if volatility <= 0.20:
        return "Low"
    elif volatility <= 0.35:
        return "Medium"
    else:
        return "High"

    companies["Risk"] = companies[
        "Historical Volatility"
    ].apply(get_risk_level)

def load_company_data():

    financials = pd.read_csv(dataset)
    company_info = pd.read_csv(company_info_file)

    df = financials.merge(
        company_info[
            [
                "Ticker",
                "SimFinId",
                "Company_Name",
                "Sector",
                "Industry",
                "Currency"
            ]
        ],
        on="Ticker",
        how="left"
    )

    return df

def get_latest_companies(df):

    df = df.sort_values(
        "Fiscal Year"
    )

    latest = (
        df
        .groupby(
            "Ticker",
            as_index=False
        )
        .tail(1)
        .copy()
    )

    return latest

def apply_preferences(
    df,
    sector,
    risk,
    goal
):
    filtered = df.copy()

#sector
    if sector != "All":
        filtered = filtered[
        filtered["Sector"] == sector
    ].copy()

# risk
    if "Historical Volatility" in filtered.columns:

        volatility_percentile = (
            filtered["Historical Volatility"]
            .rank(pct=True)
        )

        if risk == "Low":
            filtered["Preference Bonus"] = (1 - volatility_percentile) * 10

        elif risk == "Moderate":
            filtered["Preference Bonus"] = (
                1 - abs(volatility_percentile - 0.5) * 2) * 5

        else:
            filtered["Preference Bonus"] = (volatility_percentile) * 3

    else:
        filtered["Preference Bonus"] = 0

#goal
    if goal == "High Profitability":
        filtered["Preference Bonus"] += (
            normalize_score(
                filtered["Net Profit Margin"]
            ) * 0.10
        )

    elif goal == "Maximum Growth":
        filtered["Preference Bonus"] += (
            normalize_score(
                filtered["Revenue Growth"]
            ) * 0.10
        )

    elif goal == "Long Term Stability":
        filtered["Preference Bonus"] += (
            filtered[
                "Financial Stability Score"
            ] / 25 * 10
        )

    elif goal == "Fast Investment Recovery":
        filtered["Preference Bonus"] += (
            normalize_score(
                filtered["Historical Return"]
            ) * 0.10
        )

    elif goal == "Balanced Return and Risk":
        return_component = normalize_score(
            filtered["Historical Return"]
        )

        volatility_component = (
            100
            - normalize_score(
                filtered["Historical Volatility"]
            )
        )

        filtered["Preference Bonus"] += (
            (
                return_component
                + volatility_component
            ) / 2
        ) * 0.10

    filtered["Adjusted Score"] = (
        filtered["Investment Score"]
        + filtered["Preference Bonus"]
    )

    return filtered

def get_top_5(
    sector,
    risk,
    goal
):

    df = load_company_data()
    latest = get_latest_companies(df)
    latest = calculate_investment_score(latest)

    filtered = apply_preferences(
        latest,
        sector,
        risk,
        goal
    )

    companies_scanned = filtered["Ticker"].nunique()

    top_5 = (
        filtered
        .sort_values(
            "Adjusted Score",
            ascending=False
        )
        .head(5)
        .reset_index(drop=True)
    )

    top_5["Rank"] = (
        top_5.index + 1
    )

    return top_5, companies_scanned
```
