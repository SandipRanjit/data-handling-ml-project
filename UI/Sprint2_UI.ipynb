# Utils

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/utils.py

import os
import base64
from pathlib import Path

ASSET_PATH = ("/content/drive/MyDrive/data_handling_and_machine_learning/UI/assets/")

GREEN = "#6FCF97"
PURPLE = "#A05ACB"
BLUE = "#00D6FF"
YELLOW = "#f7ce46"
BLUE_PURPLE = "#7B7EDB"

def load_svg(filename):
    path = os.path.join(ASSET_PATH, filename)

    with open(path, "r") as f:
        return f.read()

def load_svg_as_image(filename):

    svg = load_svg(filename)

    svg = svg.replace(
        "currentColor",
        "#B86CFF"
    )

    encoded = base64.b64encode(
        svg.encode("utf-8")
    ).decode("utf-8")

    return f"data:image/svg+xml;base64,{encoded}"
```

# Chat


```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/chat.py

import streamlit as st
import json
import markdown
from modules import utils
from modules import base_qwen
from modules import finetuned_qwen
from modules import rag


def initialize_chat():

    if "messages" not in st.session_state:
        st.session_state.messages = [
            {
                "role": "assistant",
                "message": (
                    "Hi! I'm your InvestIQ research assistant. "
                    "Ask me anything about the analyzed companies, "
                    "risk, sectors, or your investment strategy."
                )
            }
        ]

    if "chat_method" not in st.session_state:
        st.session_state.chat_method = "Base Qwen"

def normalize_ai_response(response):
    """
    Convert model output into plain human-readable text
    for the Streamlit chat UI.
    """

    if not isinstance(response, str):
        return str(response)

    response = response.strip()

    try:
        parsed = json.loads(response)

        if isinstance(parsed, dict):

            if parsed.get("answer"):
                return str(parsed["answer"])

            if parsed.get("summary"):
                return str(parsed["summary"])

    except (json.JSONDecodeError, TypeError):
        pass

    return response

def generate_chat_response(question):

    method = st.session_state.chat_method

    if method == "Base Qwen":
        response = base_qwen.generate_response(question)

    elif method == "Fine-Tuned Qwen":
        response = finetuned_qwen.generate_response(question)

    elif method == "RAG":
        rag_result = rag.generate_response(question)
        if isinstance(rag_result, dict):
            response = rag_result.get("answer", "")
        else:
            response = rag_result

    else:
        response = "Please select an analysis method."

    return normalize_ai_response(response)


def render_chat():

    sparkles_svg = utils.load_svg("sparkles.svg")
    user_svg = utils.load_svg("user-round.svg")
    sparkles_img = utils.load_svg_as_image("sparkles.svg")

    with st.container(border=True):

        st.markdown(
            f"""
            <div class="chat-header">
                <div class="chat-title">
                    AI CHAT
                </div>
                <div class="chat-badge">
                    <span class="result-icon">
                        {sparkles_svg}
                    </span>
                    AI assistant
                </div>
            </div>
            """,
            unsafe_allow_html=True
        )

        st.session_state.chat_method = st.selectbox(
            "Analysis Method",
            [
                "Base Qwen",
                "Fine-Tuned Qwen",
                "RAG"
            ],
            index=[
                "Base Qwen",
                "Fine-Tuned Qwen",
                "RAG"
            ].index(st.session_state.chat_method),
            key="chat_method_selector"
        )

        for chat in st.session_state.messages:
            if chat["role"] == "user":
                st.markdown(
                    f"""
                      <div class="user-message">
                          <div class="user-bubble">
                              {chat["message"]}
                          </div>
                          <div class="user-icon">
                              {user_svg}
                          </div>
                      </div>
                    """,
                    unsafe_allow_html=True
                )

            else:
                formatted_message = markdown.markdown(chat["message"])

                st.html(
                    f"""
                    <div class="chat-message">
                        <div class="chat-icon">
                            <img src="{sparkles_img}" alt="AI">
                        </div>
                        <div class="chat-bubble">
                            {formatted_message}
                        </div>
                    </div>
                    """
                )

        with st.form("chat_form", clear_on_submit=True):

            chat_col1, chat_col2 = st.columns([10, 1])

            with chat_col1:

                user_question = st.text_input(
                    "Ask a question",
                    placeholder="Ask about companies, risk, sectors...",
                    label_visibility="collapsed"
                )

            with chat_col2:

                send = st.form_submit_button("➤")

        if send and user_question.strip():

            question = user_question.strip()

            st.session_state.messages.append({
                "role": "user",
                "message": question
            })

            response = generate_chat_response(question)

            st.session_state.messages.append({
                "role": "assistant",
                "message": response
            })

            st.rerun()
```

# Calculations

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/investment_score.py

import pandas as pd
import numpy as np

dataset = "/content/drive/MyDrive/data_handling_and_machine_learning/datasets/merged_dataset/merged_dataset.csv"
company_info_file = "/content/drive/MyDrive/data_handling_and_machine_learning/week_03/outputs/olga_companies_info.csv"

# missing values give a neutral score
def normalize_score(series):
    scores = series.rank(pct=True, method="average") * 100

    return scores.fillna(50)

def calculate_profitability(df):

    net_margin = normalize_score(df["Net Profit Margin"])
    operating_margin = normalize_score(df["Operating Profit Margin"])
    fcf_margin = normalize_score(df["FCF Margin"])
    raw_score = (
        net_margin * 0.40
        + operating_margin * 0.30
        + fcf_margin * 0.30
    )
    df["Profitability Score"] = (raw_score * 0.30)

    return df

def calculate_historical_growth(df):

    revenue_growth = normalize_score(df["Revenue Growth"])
    net_income_growth = normalize_score(df["Net Income Growth"])
    operating_income_growth = normalize_score(df["Operating Income Growth"])
    asset_growth = normalize_score(df["Asset Growth"])
    raw_score = (
        revenue_growth * 0.40
        + net_income_growth * 0.30
        + operating_income_growth * 0.20
        + asset_growth * 0.10
    )
    df["Historical Growth Score"] = (raw_score * 0.25)

    return df

def calculate_financial_stability(df):

    current_ratio = normalize_score(df["Current Ratio"])
    debt = 100 - normalize_score(df["Debt-to-Equity"])
    free_cash_flow = normalize_score(df["Free Cash Flow"])
    equity = normalize_score(df["Total Equity"])
    raw_score = (
        current_ratio * 0.25
        + debt * 0.30
        + free_cash_flow * 0.25
        + equity * 0.20
    )
    df["Financial Stability Score"] = (
        raw_score * 0.25
    )

    return df

def calculate_historical_return(df):

    historical_return = normalize_score(df["Historical Return"])
    low_volatility = (100 - normalize_score(df["Historical Volatility"]))
    raw_score = (historical_return * 0.70 + low_volatility * 0.30)
    df["Historical Return Score"] = (raw_score * 0.20)

    return df

def calculate_investment_score(df):

    df = df.copy()
    df = calculate_profitability(df)
    df = calculate_historical_growth(df)
    df = calculate_financial_stability(df)
    df = calculate_historical_return(df)

    df["Investment Score"] = (
        df["Profitability Score"]
        + df["Historical Growth Score"]
        + df["Financial Stability Score"]
        + df["Historical Return Score"]
    )

    return df

def get_risk_level(volatility):

    if pd.isna(volatility):
        return "Medium"
    if volatility <= 0.20:
        return "Low"
    elif volatility <= 0.35:
        return "Medium"
    else:
        return "High"

def load_company_data():

    financials = pd.read_csv(dataset)
    company_info = pd.read_csv(company_info_file)

    df = financials.merge(
        company_info[
            [
                "Ticker",
                "SimFinId",
                "Company_Name",
                "Sector",
                "Industry",
                "Currency"
            ]
        ],
        on="Ticker",
        how="left"
    )

    return df

def get_latest_companies(df):

    df = df.sort_values(
        "Fiscal Year"
    )

    latest = (
        df
        .groupby(
            "Ticker",
            as_index=False
        )
        .tail(1)
        .copy()
    )

    return latest

def apply_preferences(
    df,
    sector,
    risk,
    goal
):
    filtered = df.copy()

#sector
    if sector != "All":
        filtered = filtered[
            filtered["Sector"] == sector
        ].copy()

# risk
    volatility_percentile = normalize_score(
        filtered["Historical Volatility"]
    )

    if risk == "Low":
        risk_match = 100 - volatility_percentile

    elif risk in ["Moderate", "Medium"]:
        percentile = (
            filtered["Historical Volatility"]
            .rank(pct=True)
        )

        risk_match = (
            1 - abs(percentile - 0.5) * 2
        ) * 100

    elif risk == "High":
        risk_match = volatility_percentile

    else:
        risk_match = pd.Series(
            50,
            index=filtered.index
        )

#goal
    if goal == "High Profitability":
        goal_match = (
            normalize_score(
                filtered["Net Profit Margin"]
            ) * 0.40
            +
            normalize_score(
                filtered["Operating Profit Margin"]
            ) * 0.30
            +
            normalize_score(
                filtered["FCF Margin"]
            ) * 0.30
        )

    elif goal == "Maximum Growth":
        goal_match = (
            normalize_score(
                filtered["Revenue Growth"]
            ) * 0.40
            +
            normalize_score(
                filtered["Net Income Growth"]
            ) * 0.30
            +
            normalize_score(
                filtered["Operating Income Growth"]
            ) * 0.20
            +
            normalize_score(
                filtered["Asset Growth"]
            ) * 0.10
        )

    elif goal == "Long Term Stability":
        goal_match = (
            filtered["Financial Stability Score"]
            / 25
            * 100
        )

    elif goal == "Fast Investment Recovery":
        goal_match = normalize_score(
            filtered["Historical Return"]
        )

    elif goal == "Balanced Return and Risk":
        return_match = normalize_score(
            filtered["Historical Return"]
        )

        low_volatility_match = (
            100
            - normalize_score(
                filtered["Historical Volatility"]
            )
        )

        goal_match = (
            return_match * 0.50
            + low_volatility_match * 0.50
        )

    else:
        goal_match = pd.Series(
            50,
            index=filtered.index
        )

    filtered["Goal Match"] = goal_match
    filtered["Risk Match"] = risk_match

    filtered["Preference Score"] = (
        filtered["Goal Match"] * 0.70
        + filtered["Risk Match"] * 0.30
    )

    filtered["Ranking Score"] = (
        filtered["Investment Score"]
        + filtered["Preference Score"] * 0.25
    )

    return filtered

def get_top_5(
    sector,
    risk,
    goal
):

    df = load_company_data()
    latest = get_latest_companies(df)
    latest = calculate_investment_score(latest)

    filtered = apply_preferences(
        latest,
        sector,
        risk,
        goal
    )

    companies_scanned = filtered["Ticker"].nunique()

    top_5 = (
        filtered
        .sort_values(
            "Ranking Score",
            ascending=False
        )
        .head(5)
        .reset_index(drop=True)
    )

    top_5["Rank"] = (
        top_5.index + 1
    )

    return top_5, companies_scanned
```

# Classical ML Prediction Module

This module loads the trained Decision Tree, the saved 31-feature list, and the median imputer used during training.

Its only responsibility is to generate the probability that a company's 1-year future return will be positive.

The existing investment-ranking system, fine-tuned Qwen model, base Qwen model, and RAG system remain separate components.

```python
%%writefile /content/drive/MyDrive/data_handling_and_machine_learning/UI/modules/classical_ml.py

import json
import joblib
import pandas as pd


MODEL_PATH = ("/content/drive/MyDrive/data_handling_and_machine_learning/datasets/classical_ml_classification_dataset/InvestIQ_decision_tree_model.pkl")

FEATURES_PATH = ("/content/drive/MyDrive/data_handling_and_machine_learning/datasets/classical_ml_classification_dataset/InvestIQ_classical_ml_features.json")

IMPUTER_PATH = ("/content/drive/MyDrive/data_handling_and_machine_learning/datasets/classical_ml_classification_dataset/InvestIQ_classical_ml_imputer.pkl")

_model = None
_imputer = None
_features = None

def load_artifacts():
    global _model, _imputer, _features

    if _model is None:
        _model = joblib.load(MODEL_PATH)

    if _imputer is None:
        _imputer = joblib.load(IMPUTER_PATH)

    if _features is None:
        with open(FEATURES_PATH, "r", encoding="utf-8") as f:
            _features = json.load(f)

    return _model, _imputer, _features

def predict_positive_return_probability(company_data):
    """
    Predict the probability of a positive 1-year future return.

    company_data must contain the 31 features used during training.
    """

    model, imputer, features = load_artifacts()

    input_data = company_data[features].copy()

    input_imputed = imputer.transform(input_data)

    input_imputed = pd.DataFrame(
        input_imputed,
        columns=features,
        index=input_data.index
    )

    probability = model.predict_proba(input_imputed)[0][1]

    return float(probability)
```

# Connect the Existing RAG Pipeline to Streamlit

The existing RAG retrieval and reranking logic is preserved.

The Streamlit RAG module will:
1. Embed the user's question with MiniLM.
2. Retrieve candidates from the existing Chroma collection.
3. Rerank them with the existing Cross-Encoder.
4. Build the existing grounded RAG prompt.
5. Use the existing Fine-Tuned Qwen module for generation.

The Fine-Tuned Qwen model is not implemented or loaded separately here.

```python
rag_module_path = (
    "/content/drive/MyDrive/"
    "data_handling_and_machine_learning/"
    "UI/modules/rag.py"
)

rag_code = r'''
import chromadb
from sentence_transformers import SentenceTransformer, CrossEncoder

from modules import finetuned_qwen


CHROMA_PATH = (
    "/content/drive/MyDrive/"
    "data_handling_and_machine_learning/"
    "chroma_db"
)

COLLECTION_NAME = "investiq_rag"

_embedding_model = None
_reranker = None
_collection = None


def load_rag_components():
    global _embedding_model
    global _reranker
    global _collection

    if _embedding_model is None:
        _embedding_model = SentenceTransformer(
            "all-MiniLM-L6-v2"
        )

    if _reranker is None:
        _reranker = CrossEncoder(
            "cross-encoder/ms-marco-MiniLM-L-6-v2"
        )

    if _collection is None:
        chroma_client = chromadb.PersistentClient(
            path=CHROMA_PATH
        )

        _collection = chroma_client.get_collection(
            name=COLLECTION_NAME
        )

    return _embedding_model, _reranker, _collection


def retrieve_chunks(
    query,
    n_candidates=10,
    n_results=5,
    company=None
):
    embedding_model, reranker, collection = load_rag_components()

    query_embedding = embedding_model.encode(
        query
    ).tolist()

    if company is not None:
        results = collection.query(
            query_embeddings=[query_embedding],
            n_results=n_candidates,
            where={"company": company}
        )
    else:
        results = collection.query(
            query_embeddings=[query_embedding],
            n_results=n_candidates
        )

    candidates = []

    for document, metadata, distance in zip(
        results["documents"][0],
        results["metadatas"][0],
        results["distances"][0]
    ):
        candidates.append({
            "text": document,
            "metadata": metadata,
            "distance": distance
        })

    candidate_pairs = [
        (query, result["text"])
        for result in candidates
    ]

    scores = reranker.predict(candidate_pairs)

    reranked = []

    for result, score in zip(candidates, scores):
        reranked.append({
            "text": result["text"],
            "metadata": result["metadata"],
            "distance": result["distance"],
            "reranker_score": float(score)
        })

    reranked.sort(
        key=lambda x: x["reranker_score"],
        reverse=True
    )

    return reranked[:n_results]


def build_reranked_context(retrieved_chunks):
    context_parts = []

    for rank, result in enumerate(
        retrieved_chunks,
        start=1
    ):
        metadata = result["metadata"]

        context_parts.append(
            f"""
Source ID: {metadata["chunk_id"]}
Company: {metadata["company"]}
Section: {metadata["section"]}
Retrieval Rank: {rank}

Text:
{result["text"]}
"""
        )

    return "\n".join(context_parts)


def generate_response(question, company=None):
    reranked_results = retrieve_chunks(
        query=question,
        n_candidates=10,
        n_results=5,
        company=company
    )

    reranked_context = build_reranked_context(
        reranked_results
    )

    rag_prompt_v2 = f"""
You are an investment research assistant.

Answer the user's question using ONLY the retrieved source context below.

GROUNDING RULES:
1. Treat the retrieved context as the complete evidence boundary.
2. Use only information explicitly supported by the retrieved context.
3. Do not use outside knowledge to fill missing information.
4. If the context does not contain enough evidence to answer the question,
   explicitly state that the available sources do not provide enough information.
5. Cite a source ID only when the answer uses a factual detail supported by that source.
6. Do not cite a source merely because it was included in the retrieved context.
7. Do not cite irrelevant sources.
8. Never invent source IDs.
9. Every cited source ID must appear in the retrieved context.
10. Keep the answer concise and directly related to the user's question.

Return ONLY valid JSON using exactly this structure:

{{
  "answer": "A concise grounded answer.",
  "sources": ["source_id_1", "source_id_2"]
}}

User question:
{question}

Retrieved context:
{reranked_context}
"""

    return finetuned_qwen.generate_response(
        rag_prompt_v2
    )
'''

with open(rag_module_path, "w", encoding="utf-8") as f:
    f.write(rag_code)

print("rag.py created successfully.")
print(rag_module_path)
```

## Make RAG Source IDs Reliable

The language model can generate incorrect source IDs.

Therefore:
- Qwen generates the answer.
- Python keeps the actual retrieved source IDs.
- The application never trusts source IDs generated by Qwen.

The retrieved source metadata remains available for later source-display logic.

```python
def generate_response(question, company=None):
    reranked_results = retrieve_chunks(
        query=question,
        n_candidates=10,
        n_results=5,
        company=company
    )

    reranked_context = build_reranked_context(
        reranked_results
    )

    rag_prompt_v2 = f"""
        You are an investment research assistant.

        Answer the user's question using ONLY the retrieved source context below.

        GROUNDING RULES:
        1. Treat the retrieved context as the complete evidence boundary.
        2. Use only information explicitly supported by the retrieved context.
        3. Do not use outside knowledge.
        4. If the context does not contain enough information, say so.
        5. Do not generate source IDs.
        6. Return only the answer itself.
        7. Keep the answer concise and directly related to the user's question.

        User question:
        {question}

        Retrieved context:
        {reranked_context}
    """

    response = finetuned_qwen.generate_response(rag_prompt_v2)

    return {
        "answer": response,
        "retrieved_sources": [
            {
                "source_id": result["metadata"]["chunk_id"],
                "company": result["metadata"]["company"],
                "section": result["metadata"]["section"],
                "reranker_score": result["reranker_score"]
            }
            for result in reranked_results
        ]
    }
```
