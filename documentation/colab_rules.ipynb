# InvestIQ Team Collaboration Rules

## Purpose

These rules help us collaborate efficiently, keep the project organized, and ensure every team member learns the technologies used throughout the project.

---

## 1. Main Notebook

* Only one responsible member updates the **Main Notebook**.
* Team members should never edit the Main Notebook directly.
* All completed work will be merged into the Main Notebook after review.

---

## 2. Sandbox Notebooks

* Every member works in their own notebook inside the **sandboxes/** folder.
* Feel free to experiment, test ideas, and make mistakes in your sandbox.
* Sandbox notebooks are for learning and development only.

---

## 3. Code Integration

Before requesting code to be added to the Main Notebook:

* Make sure it runs without errors.
* Remove unnecessary test code.
* Explain briefly what your code does.
* Inform the responsible member which cells or functions should be copied.

---

## 4. Folder Management

Only one member is responsible for:

* Creating folders
* Renaming files
* Moving notebooks
* Organizing assets
* Maintaining documentation

This prevents duplicate files and accidental changes.

---

## 5. Learning Policy

The objective of this project is for **every team member to understand the complete machine learning workflow**.

Everyone should participate in learning:

* Data collection
* Data cleaning
* Dataset creation
* Fine-tuning
* Evaluation
* RAG implementation
* Streamlit prototype

No member should work on only one component throughout the project.

---

## 6. Communication

If you:

* discover a useful resource,
* fix a bug,
* improve code,
* or encounter a problem,

share it with the team before the next meeting.

---

## 7. Weekly Meetings

Each meeting should answer three questions:

1. What did I learn this week?
2. What did I complete?
3. What is blocking my progress?

---

## 8. Code Quality

Before code is merged:

* It should be readable.
* Variable names should be meaningful.
* Remove unused code.
* Add comments where necessary.

---

## 9. Documentation

Any important decision should be documented.

Examples include:

* Dataset selection
* Model selection
* Prompt changes
* Evaluation methods
* Architecture changes

---

## 10. Team Values

Throughout this project we aim to:

* Learn together.
* Help each other.
* Ask questions freely.
* Keep the project organized.
* Prioritize understanding over speed.
* Produce a project we are proud to present.
