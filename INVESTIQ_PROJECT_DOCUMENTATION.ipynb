# InvestIQ — AI Investment Analytics Platform
---

## 1. Project Overview

**InvestIQ** is an AI-assisted investment analytics platform developed as a team project for the **Data Handling and Machine Learning** course.

The project combines structured financial data, classical machine learning, a fine-tuned language model, and Retrieval-Augmented Generation (RAG) into a single interactive Streamlit application.

The purpose of InvestIQ is to help users explore companies and financial information through:

- Financial data analysis
- Company comparison and ranking
- Investment scoring
- User-defined investment preferences
- Classical machine learning
- AI-generated financial explanations
- Retrieval-Augmented Generation using company financial documents

The application is designed as a **research and analysis tool**. Its results are based on the project's financial dataset, implemented scoring methods, machine-learning model, and retrieved documents.

---

## 2. Problem Statement

Investment analysis can require the comparison of many companies and large amounts of financial information.

A user may need to:

- Compare financial performance between companies
- Examine profitability and historical growth
- Assess financial stability and risk
- Consider historical investment returns
- Filter companies according to personal investment preferences
- Understand financial information presented in company documents

The project explores how machine learning and language models can be combined with structured financial data and document retrieval to make this analysis easier to explore through one interface.

## 12. Retrieval-Augmented Generation (RAG)

The RAG component allows the system to answer questions using retrieved company financial documents rather than relying only on the language model's internal knowledge.

**RAG data**

Original financial documents are stored under:

`RAG_financial_docs/`

The documents are organized by company.

Processed RAG chunks are stored in:

`datasets/merged_chunks_RAG/
InvestIQ_all_RAG_chunks.json`

The RAG database is stored in:

`chroma_db/`

The current Chroma collection is:

`investiq_rag`

---

**RAG Process**

The RAG pipeline follows these stages:

**1. Document preparation**

Company financial documents are collected and divided into smaller chunks.

**2. Embedding**

Each chunk is converted into a numerical vector using:

`all-MiniLM-L6-v2`

**3. Vector retrieval**

The embeddings are stored in ChromaDB.

When the user asks a question, the question is embedded and compared with the stored document chunks.

**4. Candidate retrieval**

The system initially retrieves a set of potentially relevant document chunks.

**5. Reranking**

The retrieved candidates are reranked using:

`cross-encoder/ms-marco-MiniLM-L-6-v2`

This allows the system to select the most relevant pieces of retrieved context.

**6. Grounded generation**

The selected context is provided to the fine-tuned Qwen model.

The RAG prompt instructs the model to:

- Use only the retrieved context
- Avoid inventing information
- State when the available context is insufficient
- Use source IDs for supported factual information
- Avoid inventing source IDs

The RAG implementation is:

`UI/modules/rag.py`

---

## 13. RAG Source Tracking

One of the important design decisions in the project is separating retrieval evidence from the language model's response.

The Python RAG pipeline determines which document chunks were actually retrieved.

Each retrieved source contains information such as:

```
Source ID
Company
Section
Reranker score
```

The model is instructed to cite only source IDs that actually exist in the retrieved context.

This provides a clearer connection between an AI-generated answer and the financial documents used to generate it.

## 22. Known Limitations

The current implementation has several limitations.

### 22.1 Project data coverage

The quality of the analysis depends on the financial data available in the project's dataset and documents.

### 22.2 Language-model limitations

Qwen is a relatively small language model compared with larger commercial language models. Its responses may therefore be less detailed or less robust for complex financial questions.

### 22.3 RAG retrieval

The quality of a RAG answer depends on retrieving the correct and relevant document chunks.

Incorrect or incomplete retrieval can affect the generated answer.

### 22.4 Computational performance

Model loading and text generation can take noticeable time in the Google Colab environment.

The RAG method also requires additional processing for embedding and reranking.

### 22.5 Investment scoring

The investment score is a project-defined analytical framework. It should not be interpreted as a guaranteed prediction of future investment performance.

### 22.6 Not financial advice

InvestIQ is an educational and technical project demonstrating machine learning, NLP, RAG, and financial-data analysis.

Its output should not be treated as professional financial advice or a guarantee of investment performance.
